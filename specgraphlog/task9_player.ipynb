Построить нечёткое множество «игрок уверенной ИЛИ звёздной техники».

Построить множество «игрок среднего и высокого роста».

```python
# technique, weak, confident, star
rows = [
    (0,   1.00, 0,    0   ), (10,  0.75, 0,    0   ), (20,  0.50, 0,    0   ),
    (30,  0.25, 0.33, 0   ), (40,  0,    0.67, 0   ), (50,  0,    1.00, 0   ),
    (60,  0,    0.67, 0   ), (70,  0,    0.33, 0   ), (80,  0,    0,    0.33),
    (90,  0,    0,    0.67), (100, 0,    0,    1.00),
]

cols = ["technique", "weak", "confident", "star", "confident_or_star"]
table_rows = []
for technique, weak, confident, star in rows:
    confident_or_star = max(confident, star)
    table_rows.append([technique, weak, confident, star, confident_or_star])

from IPython.display import HTML, display
def fmt(v):
    return f"{v:.2f}" if isinstance(v, float) else str(v)
th = "".join(f"<th style='text-align:right;padding:4px 10px;border:1px solid #ccc;'>{c}</th>" for c in cols)
trs = "".join("<tr>" + "".join(f"<td style='text-align:right;padding:4px 10px;border:1px solid #ccc;'>{fmt(v)}</td>" for v in r) + "</tr>" for r in table_rows)
display(HTML(f"<table style='border-collapse:collapse;font-family:monospace;'><thead><tr>{th}</tr></thead><tbody>{trs}</tbody></table>"))
```

```python
# technique, weak, confident, star
rows = [
    (0,   1.00, 0,    0   ), (10,  0.75, 0,    0   ), (20,  0.50, 0,    0   ),
    (30,  0.25, 0.33, 0   ), (40,  0,    0.67, 0   ), (50,  0,    1.00, 0   ),
    (60,  0,    0.67, 0   ), (70,  0,    0.33, 0   ), (80,  0,    0,    0.33),
    (90,  0,    0,    0.67), (100, 0,    0,    1.00),
]

import matplotlib.pyplot as plt

technique = [r[0] for r in rows]
weak = [r[1] for r in rows]
confident = [r[2] for r in rows]
star = [r[3] for r in rows]
confident_or_star = [max(c, s) for c, s in zip(confident, star)]

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(technique, weak, "o-", label="weak", color="#3aa03a")
ax.plot(technique, confident, "o-", label="confident", color="#e08e2b")
ax.plot(technique, star, "o-", label="star", color="#c0392b")
ax.plot(technique, confident_or_star, "o-", label="confident_or_star", color="#2f6fb0")
ax.set_xlabel("technique")
ax.set_ylabel("mu(x)")
ax.set_title("technique terms")
ax.legend()
ax.grid(True, linestyle="--", alpha=0.4)
fig.tight_layout()
plt.show()
```
