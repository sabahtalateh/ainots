## Нечёткие (лингвистические) переменные

**Лингвистическая переменная** — переменная, значения которой не числа, а слова естественного языка: «холодно», «тепло», «горячо». Каждому такому значению (терму) сопоставляется **нечёткое множество** — функция принадлежности, которая для каждого значения физической величины задаёт степень принадлежности от 0 до 1.

В отличие от чёткого разбиения (где, например, «холодно» = строго ниже 15°), нечёткие множества **перекрываются**: одна и та же температура может частично принадлежать сразу нескольким термам.

![Нечёткие множества: холодно, тепло, горячо](images/fuzzy_temp.svg)

На графике при выбранной температуре (вертикальная линия):

- степень принадлежности к «холодно» ≈ 0,8;
- к «тепло» ≈ 0,25;
- к «горячо» = 0.

То есть значение «скорее холодно, немного тепло, совсем не горячо». Степени принадлежности показывают, насколько значение соответствует каждому терму, и не обязаны в сумме давать 1 (в отличие от вероятностей).

## Неопределённость высказываний

Зачем вообще нужна нечёткая логика — реальные знания и формулировки часто неточны:

- **Неточность знаний** — данные и правила не всегда строгие и полные;
- **применение точных методов затруднено** — классическая (чёткая) математика плохо работает с размытыми понятиями;
- **неточность (расплывчатость) формулировок** — слова естественного языка («тепло», «высокий», «быстро») не имеют чётких границ.

Нечёткая логика как раз даёт аппарат для работы с такой неопределённостью: вместо «истина/ложь» — степень истинности из диапазона [0, 1].

## Нечёткая логика (с 1965 г.)

**Нечёткая (неточная) логика** — fuzzy logic, fuzzy set theory.

- Предложена **Лотфи Заде** (Lotfi Zadeh) в 1965 году.
- Основана на интерпретации терминов естественного языка — **лингвистических переменных** (например, «высокий», «быстро», «тепло»).
- Ключевая идея: в отличие от классической (булевой) логики, где высказывание строго истинно или ложно (1 или 0), в нечёткой логике истинность может принимать любое значение из диапазона [0, 1] — степень истинности.

## Логическая переменная

Логические переменные различаются по числу допустимых значений:

- **Булева (двузначная):** только ИСТИНА или ЛОЖЬ — значения из множества {0, 1}.
- **Многозначная логика:** конечный набор значений {0, 1, 2, …, n}. Пример — «расстояние»: {близко, средне, далеко}.

**Нечёткая логика** идёт ещё дальше: истинность — это не набор дискретных уровней, а любое значение из непрерывного отрезка [0, 1]. То есть последовательность усложнения такая: булева (2 значения) → многозначная (n значений) → нечёткая (континуум значений).

## Нечёткая переменная

Нечёткая переменная задаётся **тройкой (α, X, A)**:

- **α** — наименование переменной;
- **X** — универсальное множество (область определения α);
- **A** — нечёткое множество на X, задающее ограничения на значения α через функцию принадлежности $\mu_A(x)$. То есть μ — степень уверенности, что значение x относится к нечёткой переменной α.

**Пример.** Переменная α = «возраст» с термами: юный, молодой, зрелый, пожилой, старый.

- Область определения: X = {0; 125} (лет).
- Терм «молодой»: носитель A = {15; 50}.
- Терм «зрелый»: носитель A = {30; 65}.

Степени принадлежности $\mu_A(x)$:

| Возраст x | молодой | зрелый |
|---|---|---|
| 30 | 0,8 | 0,3 |
| 45 | 0,1 | 0,75 |

Форма записи элемента нечёткого множества — «значение/принадлежность»: например, 30/0,8 (или 0,8/30).

Термы перекрываются: в 30 лет человек «скорее молодой» (0,8) и немного «зрелый» (0,3); в 45 — «скорее зрелый» (0,75) и почти не «молодой» (0,1).

## Пример: размер клиента и расчёт цены

Нечёткая переменная α = «размер клиента» с термами **крупный / средний / малый**; принадлежность зависит от объёма покупок.

**Носители термов (объём покупок):**

- крупный клиент — от 2000 до 12000;
- средний клиент — от 500 до 8000;
- малый клиент — от 100 до 2500.

Категории перекрываются, поэтому один клиент может частично относиться сразу к нескольким.

**Пример.** Клиент X купил на 2000 ед. Его степени принадлежности:

| Терм | μ |
|---|---|
| крупный | 0,05 |
| средний | 0,8 |
| малый | 0,3 |

То есть X — «в основном средний, отчасти малый, почти не крупный».

**Задача цены.** Стоимость лицензии за единицу задана диапазонами:

- крупные: $0,02 – $0,12;
- средние: $0,08 – $0,18;
- мелкие: $0,17 – $0,25.

Как рассчитать цену X, если клиент частично относится к разным категориям? Стандартный приём — **дефаззификация**: взять представительную цену каждой категории (например, середину диапазона) и усреднить с весами-принадлежностями:

$$X = \frac{\sum_i \mu_i \cdot \text{price}_i}{\sum_i \mu_i}$$

Середины диапазонов: крупный 0,07; средний 0,13; малый 0,21. Тогда:

$$X = \frac{0{,}05\cdot0{,}07 + 0{,}8\cdot0{,}13 + 0{,}3\cdot0{,}21}{0{,}05 + 0{,}8 + 0{,}3} = \frac{0{,}1705}{1{,}15} \approx 0{,}148\ \text{\$/ед.}$$

За 2000 единиц ≈ $296,5.

*(Слайд ставит вопрос; расчёт выше — один из стандартных способов, через взвешивание цен по принадлежности.)*

## Нечёткая переменная: графическое представление

Функция принадлежности $\mu_A(x)$ может быть произвольной **функциональной зависимостью** (не только треугольником или трапецией). Она показывает степень принадлежности переменной к данному нечёткому подмножеству.

![Функция принадлежности с точками перехода](images/fuzzy_transition.svg)

Здесь показана принадлежность к «крайним» возрастам (молодой или старый): μ = 1 у молодых и у старых, с провалом в середине.

**Точка перехода** — значение аргумента, где μ = 0,5 (граница «скорее принадлежит / скорее нет»). На графике точки перехода — возраст 30 и 55.

## Определение μ-функции (S-функция)

Один из стандартных способов задать функцию принадлежности — **S-функция** (плавный переход от 0 к 1). Для терма «истинно» она задаётся кусочно:

$$\mu_{\text{истинно}}(u) = \begin{cases} 0, & 0 \le u \le a, \\[4pt] 2\left(\dfrac{u-a}{1-a}\right)^2, & a < u \le \dfrac{a+1}{2}, \\[4pt] 1 - 2\left(\dfrac{u-1}{1-a}\right)^2, & \dfrac{a+1}{2} < u \le 1. \end{cases}$$

Здесь **a** — начало подъёма (левая граница). Ключевые точки:

- $\mu(a) = 0$ — ниже $a$ принадлежность нулевая;
- $\mu\!\left(\tfrac{a+1}{2}\right) = 0{,}5$ — точка перехода (стык двух парабол);
- $\mu(1) = 1$ — полная принадлежность.

Между $a$ и 1 функция гладко растёт: сначала выпуклая парабола (ускорение), после точки перехода — вогнутая (замедление).

![S-функция (a = 0,2)](images/s_function.svg)

**Интерактивно:** покрутить параметр $a$ и посмотреть, как меняется S-функция, можно на [интерактивном графике в Desmos](https://www.desmos.com/calculator/7hfjyu6tgg).

Из таких «половинок» (растущей S-функции и зеркальной убывающей) собираются более сложные функции принадлежности — например, «провал» для «молодой или старый» с предыдущего слайда.

### Конкретный пример: терм «высокий» по росту

Применим S-функцию к росту человека. Формула работает при $u \in [0, 1]$, поэтому сначала **нормируем** рост:

$$u = \frac{\text{рост} - 150}{50} \quad (150\ \text{см} \to u=0,\ \ 200\ \text{см} \to u=1).$$

Возьмём параметр **a = 0,4**. Тогда:

- **порог:** $a = 0{,}4 \Rightarrow 150 + 0{,}4\cdot 50 = 170$ см — ниже него «не высокий» (μ = 0);
- **точка перехода:** $\frac{a+1}{2} = 0{,}7 \Rightarrow 185$ см — здесь μ = 0,5;
- **полная принадлежность:** $u = 1 \Rightarrow 200$ см — μ = 1.

Подставим $a = 0{,}4$ (тогда $1-a = 0{,}6$, $\frac{a+1}{2} = 0{,}7$) в формулу:

$$\mu_{\text{высокий}}(u) = \begin{cases} 0, & 0 \le u \le 0{,}4, \\[4pt] 2\left(\dfrac{u-0{,}4}{0{,}6}\right)^2, & 0{,}4 < u \le 0{,}7, \\[4pt] 1 - 2\left(\dfrac{u-1}{0{,}6}\right)^2, & 0{,}7 < u \le 1. \end{cases}$$

Например, рост 175 см → $u = 0{,}5$, попадает во вторую ветвь: $\mu = 2\left(\frac{0{,}5-0{,}4}{0{,}6}\right)^2 = 2\cdot(0{,}167)^2 \approx 0{,}06$.

![S-функция для роста](images/fuzzy_height.svg)

| Рост | u = (рост−150)/50 | Ветвь формулы | μ(высокий) |
|---|---|---|---|
| 165 см | 0,30 | 1-я (0) | 0 |
| 170 см | 0,40 | 1-я (0) | 0 |
| 175 см | 0,50 | 2-я (растущая) | 0,06 |
| 185 см | 0,70 | стык 2-й и 3-й | 0,50 |
| 195 см | 0,90 | 3-я (замедление) | 0,94 |
| 200 см | 1,00 | 3-я | 1,00 |

Как читать: **u** — измеренный (нормированный) рост конкретного человека, **a** — выбранный порог «высокости». Человек 185 см «высокий наполовину» (0,5), 195 см — почти полностью (0,94), а 165 см — совсем не высокий (0).

**Роль параметра a:** он двигает порог. Если взять a = 0,2 (порог 160 см), кривая сместится влево и станет «щедрее» — тот же рост 175 см даст уже μ ≈ 0,28 вместо 0,06.

## Виды μ-функций (L-R представление)

Функцию принадлежности часто задают в **L-R форме** — двумя «сторонами» относительно вершины a:

$$\mu_A(x) = \begin{cases} L\!\left(\dfrac{a-x}{\alpha}\right), & x \le a, \\[4pt] R\!\left(\dfrac{x-a}{\beta}\right), & x > a. \end{cases}$$

- **a** — вершина (где μ = 1);
- **L** — левая сторона, **R** — правая сторона (функции спада от вершины);
- **α, β** — «ширина» левого и правого склонов (коэффициенты нечёткости).

Меняя функции L, R и параметры α, β, получают разные формы функции принадлежности. Типичные:

![Виды функций принадлежности](images/mu_shapes.svg)

- **колоколообразная** — гладкий пик (гауссоподобная);
- **треугольная** — линейные склоны;
- **полукруг (полуэллипс)** — с отмеченным уровнем 50%;
- **прямоугольная** — по сути чёткое множество (интервал: μ = 1 внутри, 0 снаружи).

## Распространённые варианты процедуры G

**Процедура G** порождает новые термы из базового: применяет к функции принадлежности $\mu_t(u)$ языковые модификаторы («hedges») и комбинирует нечёткие множества.

| Квантификатор | Функция принадлежности (u ∈ U) |
|---|---|
| не t | $1 - \mu_t(u)$ |
| очень t | $(\mu_t(u))^2$ |
| более-менее t | $\sqrt{\mu_t(u)}$ |
| A и B | $\min(\mu_A(x),\ \mu_B(x))$ |
| A или B | $\max(\mu_A(x),\ \mu_B(x))$ |

- **не t** — отрицание (дополнение): 1 − μ;
- **очень t** — концентрация: квадрат сужает множество (принадлежность падает быстрее, «строже»);
- **более-менее t** — растяжение: корень расширяет множество (принадлежность растёт, «мягче»);
- **A и B** — пересечение: минимум;
- **A или B** — объединение: максимум.

**Важно:** на слайде операции записаны наоборот (A и B = max, A или B = min) — это ошибка источника. По стандарту (операторы Заде) верно **A и B = min**, **A или B = max**. Интуиция: «и» (нужны оба условия) ограничено более слабым → минимум; «или» (достаточно одного) → максимум.

![Термы «температура в комнате» и их модификаторы](images/fuzzy_hedges.svg)

На графике — базовые термы **холодно / комфортно / жарко** (сплошные линии) и их модификаторы (пунктир): «не очень холодно» = 1 − холодно², «более-менее комфортно» = √комфортно, «очень жарко» = жарко². Видно, что «очень» сужает и заостряет кривую, а «более-менее» расширяет.

## Примеры G-функции (модификаторы терма «молодой»)

Базовый терм — **молодой**, x ∈ A = {15; 50}. Возьмём два значения и их принадлежности:

- $\mu_A(20)$ = 0,8;
- $\mu_A(50)$ = 0,05.

Применяем модификаторы к этим числам:

| Модификатор | Формула | x = 20 | x = 50 |
|---|---|---|---|
| молодой (база) | $\mu_A$ | 0,8 | 0,05 |
| не молодой | $1 - \mu_A$ | 0,2 | 0,95 |
| очень молодой | $(\mu_A)^2$ | 0,64 | 0,0025 |
| чересчур молодой | $(\mu_A)^3$ | 0,512 | 0,000125 |
| более-менее молодой | $\sqrt{\mu_A}$ | 0,89 | 0,22 |

Видно, как модификаторы меняют принадлежность:

- **«не»** — дополнение (переворот): чем меньше был «молодой», тем больше «не молодой»;
- **«очень»** (квадрат) и **«чересчур»** (куб) — концентрация: сужают множество, принадлежность падает тем сильнее, чем выше степень (0,8 → 0,64 → 0,512);
- **«более-менее»** (корень) — растяжение: расширяет множество, принадлежность растёт (0,8 → 0,89).

**Замечание:** на слайде для «очень молодой» при x = 50 указано 0,00025 — это опечатка. Правильно $0{,}05^2 = 0{,}0025$. (Значения проверены расчётом.)

## Лингвистическая переменная (формальное определение)

**Лингвистическая переменная (ЛП)** — это набор из пяти элементов **(β, T, X, G, M)**:

- **β** — наименование переменной (например, «возраст»);
- **T** — **терм-множество**: множество её значений-термов (наименований нечётких переменных); у каждого терма область определения — множество X. Это *базовое* терм-множество;
- **X** — универсальное множество (область определения термов);
- **G** — **синтаксическая процедура**: порождает новые термы из имеющихся. Множество T ∪ G(T) (базовые + сгенерированные термы) называется *расширенным* терм-множеством;
- **M** — **семантическая процедура**: каждому новому терму сопоставляет нечёткую переменную, то есть строит соответствующее нечёткое множество (функцию принадлежности).

**Не путать M и μ:**

- **M** — семантическая процедура (от *Meaning*, «смысл»): вход — терм, выход — нечёткое множество. Например, M(«Молодой») = функция $\mu_{\text{молодой}}(x)$;
- **μ** (греч. «мю») — функция принадлежности (*Membership*): вход — значение x, выход — степень принадлежности из [0, 1].

То есть M для каждого терма выдаёт его μ; M это правило, μ это конкретная функция.

**Соглашения (чтобы не плодить обозначения):**

- символ β используют и для названия переменной, и для её значений;
- одним символом обозначают и нечёткое множество, и его название: терм «Молодой» (значение ЛП β = «возраст») — это одновременно и нечёткое множество M(«Молодой»);
- **G** — процедура образования новых термов через связки «и», «или» и модификаторы «очень», «не», «слегка» и т.п.

Разобранные ранее темы — это две из пяти компонент: **G** (модификаторы и связки, процедура G) и **M** (функции принадлежности: S-функция и виды μ-функций).

## Пример: лингвистическая переменная «возраст»

Эксперт описывает возраст понятиями «Молодой», «Средний возраст», «Старый»; минимальный возраст 10 лет, максимальный — 120. Формализация как ЛП **(β, T, X, G, M)**:

- **β** = «возраст»;
- **T** = {«Молодой», «Средний возраст», «Старый»} — базовые термы;
- **X** = [0, 120] — область определения (годы);
- **G** — порождение новых термов через связки «и», «или» и модификаторы «очень», «не», «слегка». Например: «Молодой и средний возраст», «Не молодой»;
- **M** — сопоставляет каждому терму нечёткое множество: A₁ = «Молодой», A₂ = «Средний возраст», A₃ = «Старый», а также нечёткие множества для термов, порождённых G.

Запись **M(A₁) = [14; 50]** задаёт носитель терма «Молодой» — интервал, где принадлежность больше нуля. Полное нечёткое множество — это пары «значение + степень принадлежности»:

$$A_1 = \{\, (x,\ \mu(x)) : x \in X \,\}, \quad \mu \in [0, 1].$$

μ — степень уверенности (принадлежности), что возраст x относится к «молодой»: например, μ(20) ≈ 1, μ(35) ≈ 0,6, μ(48) ≈ 0,1, вне [14, 50] μ = 0. Помимо носителя нужна функция принадлежности на нём.

Степень уверенности μ — это не вероятность: значения μ разных термов в сумме не обязаны давать 1 (возраст 30 может относиться к «молодой» на 0,8 и к «среднему» на 0,5 одновременно).

## Нечёткое множество

**Нечёткое множество** (размытое, расплывчатое, туманное) — понятие, введённое Лотфи Заде в 1965 году (статья «Fuzzy Sets», журнал *Information and Control*). Ключевое отличие от обычного (чёткого) множества: характеристическая функция (функция принадлежности) может принимать **любые значения из [0, 1]**, а не только 0 или 1.

- **чёткое множество:** элемент либо принадлежит (1), либо нет (0);
- **нечёткое множество:** элемент принадлежит с некоторой степенью μ ∈ [0, 1].

Запись (нотация Заде):

$$A = \{\, \mu_A(x)\, /\, x \,\},$$

где $\mu_A(x)$ — функция принадлежности, принимающая значения в упорядоченном множестве **M** (обычно M = [0, 1]).

Запись $\mu_A(x)$/x читается как «элемент x со степенью принадлежности $\mu_A(x)$» — косая черта здесь **не деление**, а разделитель пары «принадлежность / элемент».

**Про обозначение M:** на этом слайде M — это множество значений функции принадлежности (обычно [0, 1]). Его не следует путать с M из определения лингвистической переменной, где M — семантическая процедура. Это разные объекты, обозначенные одной буквой.

## Примеры записи нечёткого множества

Пусть носитель T = {x₁, x₂, x₃, x₄, x₅}, множество значений M = [0, 1], и A — нечёткое множество с принадлежностями:

$\mu_A(x_1)$ = 0,3; $\mu_A(x_2)$ = 0; $\mu_A(x_3)$ = 1; $\mu_A(x_4)$ = 0,5; $\mu_A(x_5)$ = 0,9.

Одно и то же множество A записывают несколькими способами:

1. **через точку с запятой:** A = {0,3/x₁; 0/x₂; 1/x₃; 0,5/x₄; 0,9/x₅};
2. **«суммой»:** A = {0,3/x₁ + 0/x₂ + 1/x₃ + 0,5/x₄ + 0,9/x₅};
3. **таблицей:** верхняя строка — элементы, нижняя — их принадлежности:

| x₁ | x₂ | x₃ | x₄ | x₅ |
|---|---|---|---|---|
| 0,3 | 0 | 1 | 0,5 | 0,9 |

Ни «/», ни «+» здесь не означают деление и сложение — это разделитель пары «степень / элемент» и перечисление этих пар.

**Пример с возрастами.** Два нечётких множества — термы «молодой» и «средний»:

| A₁ (молодой) | 12 | 18 | 25 | 35 | 49 |
|---|---|---|---|---|---|
| μ | 1 | 0,95 | 0,85 | 0,5 | 0,1 |

| A₂ (средний) | 25 | 30 | 35 | 40 | 55 |
|---|---|---|---|---|---|
| μ | 0,05 | 0,2 | 0,35 | 0,8 | 0,7 |

Термы перекрываются: возраст 35 принадлежит «молодой» на 0,5 и «средний» на 0,35 одновременно.

## Пример (продолжение): три терма возраста

Три нечётких множества-терма лингвистической переменной «возраст», заданные кусочно-линейными (трапециевидными) функциями принадлежности на диапазоне x ∈ [10, 80]:

- **A₁ = «Молодой»** — трапеция слева (μ = 1 у молодых, спадает к среднему возрасту);
- **A₂ = «Средний возраст»** — треугольник в центре;
- **A₃ = «Старый»** — трапеция справа (растёт к старости, μ = 1 у старых).

![Термы возраста: молодой, средний, старый](images/fuzzy_age3.svg)

Функция перехода $\mu_A(x)$ взята **линейной** — для простоты (прямые склоны вместо кривых S-функций).

Термы перекрываются: в точках пересечения μ = 0,5 — там возраст «наполовину молодой, наполовину средний» (и аналогично между средним и старым). Соседние термы вместе покрывают весь диапазон возрастов.

## Нечёткие высказывания и отношения

**Нечёткое высказывание β** — высказывание, истинность которого оценивается не «истина/ложь», а числом (степенью уверенности) из некоторого интервала И. Бывает трёх видов:

1. **базовое** — высказывание A, степень уверенности которого — число из интервала И;
2. **модифицированное** — исходное высказывание β с модификаторами m («очень», «около», «более или менее» и т.п.);
3. **составное** — образованное из высказываний βᵢ, βⱼ операциями И, ИЛИ, ЕСЛИ…ТО.

**Вид нечёткой логики зависит от интервала истинностных значений И:**

| Логика | Интервал И |
|---|---|
| Логика Заде | [0; 1] |
| Логика антонимов | [0; +∞) |
| Бесконечнозначная логика | (−∞; +∞) |

Логика Заде (обычная нечёткая) работает со значениями из [0, 1]; логика антонимов и бесконечнозначная логика расширяют диапазон истинности.

*(Обозначения интервалов: на слайде европейский стиль — «[0; ∞[» это [0, +∞), «]−∞; ∞[» это (−∞, +∞).)*

## Сравнение нечётких множеств

Пусть A и B — нечёткие множества, заданные на универсальном множестве X.

- **A содержится в B**, если для любого элемента из X функция его принадлежности множеству A будет принимать значение меньшее либо равное, чем функция принадлежности множеству B:

$$A \subset B \Leftrightarrow \forall x \in X:\ \mu_A(x) \leqslant \mu_B(x).$$

- В случае, если условие $\mu_A(x) \leqslant \mu_B(x)$ выполняется не для всех $x \in X$, говорят о **степени включения** нечёткого множества A в B, которое определяется так:

$$l(A \subset B) = \min_{x \in T} \mu_B(x), \quad \text{где } T = \{x \in X;\ \mu_A(x) \leqslant \mu_B(x),\ \mu_A(x) > 0\}.$$

- Два множества называются **равными**, если они содержатся друг в друге:

$$A = B \Leftrightarrow \forall x \in X:\ \mu_A(x) = \mu_B(x).$$

- В случае, если значения функций принадлежности $\mu_A(x)$ и $\mu_B(x)$ почти равны между собой, говорят о **степени равенства** нечётких множеств A и B, например, в виде

$$E(A = B) = 1 - \max_{x \in T} |\mu_A(x) - \mu_B(x)|, \quad \text{где } T = \{x \in X;\ \mu_A(x) \neq \mu_B(x)\}.$$

Интуиция: включение $A \subset B$ — это «A нигде не превышает B» по степени принадлежности; равенство — частный случай взаимного включения ($A \subset B$ и $B \subset A$). Степени включения и равенства нужны, когда строгое условие «для всех x» не выполняется — они дают числовую меру того, насколько близко множества к отношению включения/равенства (чем ближе к 1, тем точнее выполняется условие).

## Пример: полное включение (графики не пересекаются)

Термы «молодой» — версия А и подправленная версия Б′, где Б′ нигде не опускается ниже А:

| x (возраст) | 15 | 18 | 25 | 30 | 35 | 40 |
|---|---|---|---|---|---|---|
| μ_A | 0,1 | 0,8 | 0,75 | 0,8 | 0,25 | 0 |
| μ_Б′ | 0,1 | 0,9 | 0,8 | 0,85 | 0,6 | 0,45 |

![Графики μ_A и μ_Б′: Б′ всюду не ниже А, линии нигде не пересекаются](images/fuzzy_inclusion_nocross.svg)

В каждой точке $\mu_A(x) \leqslant \mu_{Б′}(x)$ (0,1≤0,1; 0,8≤0,9; 0,75≤0,8; 0,8≤0,85; 0,25≤0,6; 0≤0,45) — условие выполняется **везде**, синяя линия нигде не выходит за пределы зелёной. Это и есть полное включение **A ⊂ Б′**: раз пересечений нет вообще, степень включения не нужна — она автоматически равна 1.

**Правило для графика:** если линия μ_A целиком лежит не выше линии μ_B (может касаться, но нигде не выступает над ней) — A ⊂ B. Если линии пересекаются хотя бы раз — строгого включения нет, и вместо да/нет считают степень включения l(A⊂B) — см. следующий пример.

## Пример включения нечётких множеств

Два варианта задания терма **«молодой»** — по версии А и по версии Б (например, два эксперта по-разному оценили принадлежность одних и тех же возрастов):

**«Молодой» по версии А:**

| x (возраст) | 15 | 18 | 25 | 30 | 35 | 40 |
|---|---|---|---|---|---|---|
| μ_A | 0,1 | 0,8 | 0,75 | 0,8 | 0,25 | 0 |

**«Молодой» по версии Б:**

| x (возраст) | 15 | 20 | 25 | 30 | 35 | 40 |
|---|---|---|---|---|---|---|
| μ_B | 0,1 | 0,9 | 0,6 | 0,7 | 0,6 | 0,45 |

![Графики μ_A и μ_B: пересекаются в области x∈[22,33], где A превосходит B](images/fuzzy_inclusion_example.svg)

На графике видно: почти всюду синяя линия (A) лежит не выше оранжевой (B) — кроме закрашенной области (x = 25 и x = 30), где A ненадолго поднимается **над** B. Именно это пересечение линий и есть визуальный признак того, что строгого включения A ⊂ B нет.

**Задача:** найти степень включения $l(A \subset B)$.

Сравниваем $\mu_A(x)$ и $\mu_B(x)$ поточечно:

| x | 15 | 18/20 | 25 | 30 | 35 | 40 |
|---|---|---|---|---|---|---|
| μ_A ≤ μ_B ? | 0,1≤0,1 ✓ | 0,8≤0,9 ✓ | 0,75≤0,6 ✗ | 0,8≤0,7 ✗ | 0,25≤0,6 ✓ | 0≤0,45 ✓ |

Условие нарушается в двух точках (x = 25 и x = 30, где A превосходит B) — значит, строгого включения A ⊂ B нет. Считаем **степень включения**:

$$T = \{15,\ 18/20,\ 35\} \quad \text{(точки, где условие выполняется и } \mu_A > 0\text{;}\ x=40\text{ исключена, там } \mu_A=0\text{)},$$

$$l(A \subset B) = \min_{x \in T} \mu_B(x) = \min(0{,}1;\ 0{,}9;\ 0{,}6) = 0{,}1.$$

**Ответ: l(A⊂B) = 0,1.** Степень низкая — её задаёт самая слабая из «прошедших» точек (x = 15, где обе версии совпадают на минимальном уровне 0,1), несмотря на то что на большинстве остальных точек B заметно «шире» A.

## Пример: степень равенства нечётких множеств

Ещё одна независимая оценка терма **«молодой»** — версия В (третий эксперт), сравним её с версией А:

| x (возраст) | 15 | 18 | 25 | 30 | 35 | 40 |
|---|---|---|---|---|---|---|
| μ_A | 0,1 | 0,8 | 0,75 | 0,8 | 0,25 | 0 |
| μ_В | 0,15 | 0,85 | 0,7 | 0,75 | 0,35 | 0,1 |

![Графики μ_A и μ_В: почти совпадают, максимальный разрыв в x=35 и x=40](images/fuzzy_equality_example.svg)

На графике кривые идут почти вплотную друг к другу — версии А и В расходятся, но несильно. Точного равенства нет ни в одной точке, поэтому считаем **степень равенства**:

$$T = \{15,\ 18,\ 25,\ 30,\ 35,\ 40\} \quad \text{(все точки, значения нигде не совпадают в точности)},$$

$$|\mu_A(x) - \mu_В(x)|: \quad 0{,}05;\ \ 0{,}05;\ \ 0{,}05;\ \ 0{,}05;\ \ 0{,}10;\ \ 0{,}10,$$

$$E(A = В) = 1 - \max_{x \in T} |\mu_A(x) - \mu_В(x)| = 1 - 0{,}10 = 0{,}90.$$

**Ответ: E(A=В) = 0,90.** На большинстве точек расхождение маленькое (0,05), но в x = 35 и x = 40 оно вырастает до 0,10 — а поскольку в формуле стоит **максимум**, именно эти две точки и определяют итоговую оценку: множества равны на 90%.

## Операции над нечёткими множествами

**Сумма (объединение)** нечётких множеств A и B — нечёткое множество, определённое функцией принадлежности:

$$\mu_{A\cup B}(x) = \mu_A(x) \vee \mu_B(x) = \max\big(\mu_A(x),\ \mu_B(x)\big).$$

**Объединение**: создаётся новое множество из элементов исходных множеств, причём для одинаковых элементов принадлежность берётся максимальной.

**Пересечением** нечётких множеств называется нечёткое множество с функцией принадлежности:

$$\mu_{A\cap B}(x) = \mu_A(x) \wedge \mu_B(x) = \min\big(\mu_A(x),\ \mu_B(x)\big).$$

**Пересечение**: создаётся новое множество из одинаковых элементов исходных множеств, принадлежность которых берётся минимальной.

Символы ∨ и ∧ здесь — это операции max и min (а не логические «или»/«и» в булевом смысле). Заодно эти формулы — те же самые операторы Заде, что уже встречались в **процедуре G** (объединение термов «A или B» = max, пересечение «A и B» = min): там источник перепутал их местами, а здесь на слайде всё указано верно.

## Пример: объединение (ИЛИ) двух нечётких множеств

Термы «молодой» по версии А и по версии Б заданы на разных наборах точек:

| x (возраст) | 15 | 18 | 20 | 25 | 30 | 35 | 40 |
|---|---|---|---|---|---|---|---|
| μ_A | 0,1 | — | 0,8 | 0,75 | 0,3 | 0,25 | — |
| μ_B | — | 0,1 | 0,75 | 0,8 | 0,7 | 0,6 | 0,4 |
| μ_{A∪B} = max | 0,1 | 0,1 | 0,8 | 0,8 | 0,7 | 0,6 | 0,4 |

«—» — точка не задана в исходном множестве (там неявно μ = 0). Объединение считаем по формуле $\mu_{A\cup B}(x) = \max(\mu_A(x), \mu_B(x))$:

- x = 15, 18 — точка есть только в одном множестве, объединение просто копирует её значение (max с нулём);
- x = 20 — max(0,8; 0,75) = 0,8 (берём A, оно больше);
- x = 25 — max(0,75; 0,8) = 0,8 (берём Б);
- x = 30 — max(0,3; 0,7) = 0,7 (берём Б);
- x = 35 — max(0,25; 0,6) = 0,6 (берём Б);
- x = 40 — только в Б, копируем 0,4.

![Объединение A∪B как верхняя огибающая графиков μ_A и μ_B](images/fuzzy_union_example.svg)

На графике красная пунктирная линия (A∪B) — это **верхняя огибающая**: в каждой точке она идёт по той из двух кривых, что выше. Так объединение выглядит графически при любых A и B.

## Пример: пересечение (И) двух нечётких множеств

Те же множества, что и в примере объединения:

| x (возраст) | 15 | 18 | 20 | 25 | 30 | 35 | 40 |
|---|---|---|---|---|---|---|---|
| μ_A | 0,1 | — | 0,8 | 0,75 | 0,3 | 0,25 | — |
| μ_B | — | 0,1 | 0,75 | 0,8 | 0,7 | 0,6 | 0,4 |
| μ_{A∩B} = min | — | — | 0,75 | 0,75 | 0,3 | 0,25 | — |

Пересечение считаем по формуле $\mu_{A\cap B}(x) = \min(\mu_A(x), \mu_B(x))$:

- x = 15, 18, 40 — точка задана только в одном множестве, во втором её нет (μ = 0), значит min = 0 — пересечение там **не определено** (в отличие от объединения, здесь такие точки выпадают, а не копируются);
- x = 20 — min(0,8; 0,75) = 0,75;
- x = 25 — min(0,75; 0,8) = 0,75;
- x = 30 — min(0,3; 0,7) = 0,3;
- x = 35 — min(0,25; 0,6) = 0,25.

![Пересечение A∩B как нижняя огибающая графиков μ_A и μ_B, определена только там, где заданы оба множества](images/fuzzy_intersection_example.svg)

На графике зелёная линия (A∩B) — **нижняя огибающая**, но только на общей области [20, 35], где заданы и A, и Б. За её пределами (x=15, 18, 40) пересечение отсутствует — там нет одного из множеств, поэтому и общей части быть не может.

## Пример области применения

Где на практике пригождаются операции над нечёткими множествами:

- **выбор показателей** — например, показателей качества, которые удовлетворят разные группы покупателей;
- **объединение** имеет значение, так как показывает возможность интегрированного влияния нескольких процессов на конечный результат (ср. IDEF0 — декомпозиция процесса на подпроцессы, чей суммарный эффект и даёт итог);
- **пересечение множеств** может быть, когда в интернет-магазине добавляешь товары в сравнение друг с другом и смотришь по параметрам их отличия — общая (пересекающаяся) часть характеристик и есть то, что можно сравнить напрямую.

## Разность нечётких множеств

**Разность**: новое множество состоит из одинаковых элементов исходных множеств.

| x (возраст) | 15 | 18 | 20 | 25 | 30 | 35 | 40 |
|---|---|---|---|---|---|---|---|
| μ_A | 0,1 | — | 0,8 | 0,75 | 0,3 | 0,25 | — |
| μ_B | — | 0,1 | 0,8 | 0,75 | 0,7 | 0,6 | 0,4 |
| μ_{A−B} | — | — | 0,8 | 0,75 | — | — | — |

В разность попадают только точки, где A и Б совпадают целиком — и по x, и по значению μ: x = 20 (0,8 = 0,8) и x = 25 (0,75 = 0,75). На остальных общих точках значения A и Б разошлись (x=30: 0,3 против 0,7; x=35: 0,25 против 0,6), поэтому туда они не попали.

**Ответ: A − B = {0,8/20;  0,75/25}.**

## Дополнение (отрицание) нечёткого множества

**Дополнение:** $\mu_{A'}(x) = 1 - \mu_A(x)$.

| x (возраст) | 15 | 20 | 25 | 30 | 35 |
|---|---|---|---|---|---|
| μ_A («молодой») | 0,1 | 0,8 | 0,75 | 0,3 | 0,25 |
| μ_A′ (дополнение к А) | 0,9 | 0,2 | 0,25 | 0,7 | 0,75 |

Каждое значение просто вычитается из 1: 1−0,1=0,9; 1−0,8=0,2; 1−0,75=0,25; 1−0,3=0,7; 1−0,25=0,75. Это та же формула, что уже встречалась в модификаторе «не» (процедура G) — здесь она применена как самостоятельная операция над множеством, а не как языковой модификатор терма.

![Графики μ_A и его дополнения μ_A′ — зеркальны относительно уровня 0,5](images/fuzzy_complement_example.svg)

На графике видно зеркальное отражение относительно горизонтали μ = 0,5: там, где A высоко (x=20,25), дополнение низкое, и наоборот (x=15,30,35). Линии обязательно пересекаются на уровне 0,5 (там μ_A = μ_A′ = 0,5) — как раз между x=15–20 и между x=25–30.

## Графическое представление операций

Все четыре операции сразу — на паре непрерывных функций принадлежности (A — трапеция, B — треугольник), а не на дискретных точках, как в примерах выше:

![Слева направо: μ_A (трапеция), μ_B (треугольник), μ_A∩B = min, μ_A∪B = max, μ¬A = 1−μ_A](images/fuzzy_ops_row.svg)

- **μ_A∩B** (пересечение) — ненулевая только там, где перекрываются носители A и B, и не выше меньшей из двух кривых;
- **μ_A∪B** (объединение) — идёт по верхней огибающей; там, где растущий край B ненадолго обгоняет ещё не поднявшийся край A, на графике виден небольшой «зубец»;
- **μ¬A** (инверсия/отрицание) — зеркальное отражение A относительно μ = 0,5, ровно 1 − μ_A в каждой точке.

## Нечёткий вывод

**Нечётким логическим выводом** называется аппроксимация зависимости каждой выходной нечёткой переменной от множества входных нечётких переменных и получение заключения в виде нечёткого множества.

Нечёткий вывод производится на основе композиционного правила. Базовым является **композиционное правило Заде**.

*Примечание. Также существуют правила Мамдани, Лукасевича и др.*

## Пример отличий логик Заде, Мамдани, Лукасевича и пр.

«Объединение» нечётких множеств можно определять не только через max — max (операторы Заде) это лишь один из вариантов. Другие распространённые варианты объединения T(a, b):

- объединение по Заде: $T(a,b) = \max(a, b)$;
- вероятностное объединение: $T(a,b) = a + b - ab$;
- объединение по Лукасевичу: $T(a,b) = \min(a+b,\ 1)$.

Примеры выполнения объединения нечётких множеств с использованием этих операций — на графике ниже (два множества A и B заданы как треугольные функции принадлежности с разными вершинами):

![Сравнение объединения по Заде, вероятностного и по Лукасевичу на паре нечётких множеств A и B](images/fuzzy_tconorms.svg)

Во всех точках выполняется порядок $\max(a,b) \leqslant a+b-ab \leqslant \min(a+b,1)$: объединение по Заде (толстая синяя линия) — самое «узкое» и в точности повторяет верхнюю огибающую A и B; вероятностное объединение (красный штрихпунктир) немного сглаживает провал между пиками; объединение по Лукасевичу (чёрный пунктир) — самое «щедрое», быстро выходит на насыщение (μ=1) там, где A и B вместе достаточно велики.

## Система нечёткого вывода

**Система нечёткого вывода** — это процесс получения нечётких заключений о требуемом управлении объектом на основе нечётких условий или предпосылок, представляющих собой информацию о текущем состоянии объекта.

![Схема системы нечёткого вывода: X → фазификатор → нечёткий логический вывод (с базой правил) → дефазификатор → Y](images/fuzzy_inference_system.svg)

Четыре блока конвейера:

1. **Фазификатор** (приведение к нечёткости) — переводит чёткое входное значение X в степени принадлежности термам входных лингвистических переменных;
2. **База правил** — набор нечётких правил «ЕСЛИ ... ТО ...», задающих зависимость выхода от входа;
3. **Нечёткий логический вывод** — применяет правила из базы к фазифицированному входу и получает нечёткое заключение (см. композиционное правило Заде выше);
4. **Дефазификатор** (приведение к чёткости) — превращает нечёткое заключение обратно в одно чёткое число Y (например, взвешенным усреднением — как в примере с ценой для клиента ранее в этом конспекте).

## Пример: полный цикл нечёткого вывода (регулятор обогревателя)

Переменная X = «температура в комнате» (термы холодно/тепло/горячо), выход Y = «мощность обогревателя», 0–100%, с термами **слабая / средняя / сильная**.

**База правил** (по одному правилу на входной терм):

1. ЕСЛИ температура **холодно**, ТО мощность **сильная**;
2. ЕСЛИ температура **тепло**, ТО мощность **средняя**;
3. ЕСЛИ температура **горячо**, ТО мощность **слабая**.

**Шаг 1. Фазификация.** Зададим термы X явно, кусочно-линейными функциями принадлежности (в °C):

$$\mu_{\text{холодно}}(x) = \begin{cases} 1, & x \le 10 \\ \dfrac{20-x}{10}, & 10 < x \le 20 \\ 0, & x > 20 \end{cases} \qquad \mu_{\text{тепло}}(x) = \begin{cases} 0, & x \le 10 \\ \dfrac{x-10}{8}, & 10 < x \le 18 \\ 1, & 18 < x \le 22 \\ \dfrac{30-x}{8}, & 22 < x \le 30 \\ 0, & x > 30 \end{cases} \qquad \mu_{\text{горячо}}(x) = \begin{cases} 0, & x \le 22 \\ \dfrac{x-22}{8}, & 22 < x \le 30 \\ 1, & x > 30 \end{cases}$$

Берём конкретную входную температуру **X = 12°C** и подставляем в каждую формулу:

- $\mu_{\text{холодно}}(12)$: попадает во вторую ветвь ($10 < 12 \le 20$) → $\frac{20-12}{10} = \frac{8}{10} = 0{,}8$;
- $\mu_{\text{тепло}}(12)$: попадает во вторую ветвь ($10 < 12 \le 18$) → $\frac{12-10}{8} = \frac{2}{8} = 0{,}25$;
- $\mu_{\text{горячо}}(12)$: $12 \le 22$ → первая ветвь → $0$.

![Термы холодно/тепло/горячо с отмеченной точкой X=12°C: μ_холодно=0,8; μ_тепло=0,25; μ_горячо=0](images/fuzzy_fuzzification_example.svg)

Итог фазификации: **μ_холодно = 0,8; μ_тепло = 0,25; μ_горячо = 0** — при 12°C в комнате «в основном холодно, немного тепло, и точно не жарко». Именно эти три числа идут дальше в правила.

**Шаг 2. Нечёткий вывод.** Каждое правило срабатывает со «силой», равной μ его условия, и на эту величину отсекает (обрезает по высоте) терм вывода:

| Правило | Условие сработало на | Выход отсекается на уровне |
|---|---|---|
| холодно → сильная | μ_холодно = 0,8 | сильная ∧ 0,8 |
| тепло → средняя | μ_тепло = 0,25 | средняя ∧ 0,25 |
| горячо → слабая | μ_горячо = 0 | слабая ∧ 0 = не активна |

Отсечённые термы объединяются (max) в одно результирующее нечёткое множество — это и есть «нечёткий логический вывод» на схеме.

![Термы слабая/средняя/сильная (пунктир), отсечённые по силе правил и объединённые (сплошная синяя область), дефазификация даёт Y≈72,9%](images/fuzzy_inference_example.svg)

**Шаг 3. Дефазификация.** Превращаем нечёткий результат в одно число — упрощённо, как и в примере с ценой для клиента: берём представительное значение каждого терма (слабая≈20%, средняя≈50%, сильная≈80%) и усредняем с весами-активациями:

$$Y = \frac{0{,}8\cdot 80 + 0{,}25\cdot 50 + 0\cdot 20}{0{,}8 + 0{,}25 + 0} = \frac{76{,}5}{1{,}05} \approx 72{,}9\%.$$

**Итог:** при температуре 12°C («в основном холодно», μ=0,8, и лишь немного «тепло», μ=0,25) обогреватель включается почти на полную мощность — 72,9%. Именно так один и тот же контур «X → фазификация → правила → дефазификация → Y» превращает нечёткие термы в конкретное управляющее число.

## Продукционное (нечёткое) правило

**Нечёткое продукционное (композиционное) правило** — это выражение вида:

$$(i):\ Q;\ P;\ A \Rightarrow B;\ S,\ F,\ N,$$

где:

- **(i)** — имя нечёткой продукции;
- **Q** — сфера применения нечёткой продукции;
- **P** — условие применимости ядра нечёткой продукции;
- **A⇒B** — **ядро** нечёткой продукции, в котором A — условие ядра (или **антецедент**), B — заключение ядра (или **консеквент**), ⇒ — знак логической секвенции (следования);
- **S** — метод или способ определения количественного значения степени истинности заключения ядра;
- **F** — коэффициент определённости или уверенности нечёткой продукции;
- **N** — постусловия продукции (не обязательны).

По сути, каждое правило из «базы правил» в схеме нечёткого вывода (например, «ЕСЛИ температура холодно ТО мощность сильная» из примера выше) — это ядро A⇒B такой продукции; остальные компоненты (Q, P, S, F, N) — дополнительные атрибуты, уточняющие область и достоверность применения правила.

## Пример импликации

**Правило:** «ЕСЛИ „оценка“ положительная, ТО „результат экзамена“ = „зачёт“, ставим в ведомость».

Разложим его по компонентам продукционного правила $(i):Q;P;A\Rightarrow B;S,F,N$:

- **Q** — образование (скрытая, неявная компонента — область, в которой действует правило);
- **P** — условие «ЕСЛИ»;
- **A** (антецедент) — «оценка»;
- **B** (консеквент) — «результат»;
- **S** — «положительная» (условие на антецедент, определяющее срабатывание ядра);
- **N** — постусловие — «ставим в ведомость»;
- **F** — определяется по μ-функции (не задаётся числом заранее, а вычисляется через степень принадлежности оценки терму «положительная»).

![μ-функция степени уверенности F от оценки: растёт от 0 при оценке 1 до 1 при оценке 5](images/fuzzy_implication_example.svg)

Чем выше оценка, тем увереннее (F ближе к 1) правило признаёт её «положительной» и тем сильнее срабатывает заключение B. Это тот же принцип, что и в правилах базы правил регулятора выше: F здесь играет роль «силы срабатывания» — как μ_холодно = 0,8 определяла, насколько сильно отсекается терм «сильная» на выходе.

## Вывод правила. Импликация «ЕСЛИ..., ТО...»

Общий вид простого правила — через уже знакомую запись «лингвистическая переменная β ЕСТЬ терм α»:

$$\text{ПРАВИЛО}\ \langle\#\rangle:\ \text{ЕСЛИ}\ \beta_1\ \text{ЕСТЬ}\ \alpha_1\ \text{ТО}\ \beta_2\ \text{ЕСТЬ}\ \alpha_2.$$

Антецедент и консеквент ядра нечёткой продукции могут быть **сложными** — состоящими из нескольких условий, соединённых связками «И», «ИЛИ», «НЕ», например:

$$\text{ПРАВИЛО}\ \langle\#\rangle:\ \text{ЕСЛИ}\ \big(\beta_1\ \text{ЕСТЬ}\ \alpha\big)\ \text{И}\ \big(\beta_2\ \text{ЕСТЬ НЕ}\ \alpha\big)\ \text{ТО}\ \big(\beta_1\ \text{ЕСТЬ НЕ}\ \beta_2\big).$$

Названия для этих связок стандартные, из логики:

- операция логического **ИЛИ** называется **дизъюнкцией**;
- операция логического **И** называется **конъюнкцией**.

**Пример** (продолжение примера с оценкой): ЕСЛИ «оценка положительная» И «посещаемость удовлетворительная ИЛИ хорошая», ТО результат «курс зачтён» — поставить в ведомость.

Здесь антецедент составной: (оценка ЕСТЬ положительная) И ((посещаемость ЕСТЬ удовлетворительная) ИЛИ (посещаемость ЕСТЬ хорошая)). Степень истинности всего антецедента считается через уже знакомые операции над степенями принадлежности — конъюнкция как min, дизъюнкция как max (операторы Заде, см. «Операции над нечёткими множествами» выше): правило сработает настолько, насколько выполнены оба условия сразу (min по «И»), при этом «посещаемость» достаточно любой из двух term'ов (max по «ИЛИ»).

## Пример: от «оценка + посещаемость» к степени уверенности в итоговой оценке

Продолжение примера с составным правилом «ЕСЛИ оценка положительная И (посещаемость удовл. ИЛИ хорошая)». Возьмём пары входных значений (оценка от 1 до 5, и соответствующая ей посещаемость) вместе со степенями принадлежности их термам:

| Оценка | Неуд. | Удовл. | Посещ. | Неуд | Удовл. | Хорошая | 3 | 4 | 5 |
|---|---|---|---|---|---|---|---|---|---|
| 1 | 1 | 0 | 0 % | 1 | 0 | 0 | | | |
| 2 | 1 | 0,1 | 20 % | 1 | 0,05 | 0 | | | |
| 3 | 0,2 | 0,5 | 50 % | 0,05 | 0,9 | 0,1 | | | |
| 4 | 0 | 0,9 | 80 % | 0 | 0,8 | 0,7 | | | |
| 5 | 0 | 1 | 100 % | 0 | 0 | 1 | | | |

Столбцы **3 / 4 / 5** — степень уверенности, что по совокупности (оценка + посещаемость) будет выставлена именно эта итоговая оценка. Заполняются построчно ниже, через базу правил, комбинирующую термы «оценка» и «посещаемость».

### База правил «оценка + посещаемость → итог» (предложенная, не из источника)

Готовой базы правил для этой комбинации в исходных материалах курса не было. Ниже — соответствие, предложенное как реконструкция (по принципу «чем лучше оба входа, тем выше итог»), а не выписанное со слайда. Если найдётся официальная таблица с лекции — её стоит использовать вместо этой.

1. ЕСЛИ оценка **Неуд.** И посещаемость **Неуд** ТО итог = **3**
2. ЕСЛИ оценка **Неуд.** И посещаемость **Удовл.** ТО итог = **3**
3. ЕСЛИ оценка **Неуд.** И посещаемость **Хорошая** ТО итог = **4**
4. ЕСЛИ оценка **Удовл.** И посещаемость **Неуд** ТО итог = **3**
5. ЕСЛИ оценка **Удовл.** И посещаемость **Удовл.** ТО итог = **4**
6. ЕСЛИ оценка **Удовл.** И посещаемость **Хорошая** ТО итог = **5**

### Что даёт эта таблица: оценка степени уверенности в высказываниях

Имея две лукап-таблицы термов (слева — для «оценка», справа — для «посещаемость») и базу правил, связывающую их с итоговой оценкой, можно оценить степень уверенности $F$ для произвольного нечёткого высказывания вида «человеку с такой-то оценкой и такой-то посещаемостью будет поставлена такая-то итоговая оценка» — не только для пар из таблицы (где оценка и посещаемость взяты из одной строки), а для любой комбинации значений.

**Высказывание 1:** «Человеку с оценкой 3 и посещаемостью 80 % будет поставлена (в зачётку) итоговая оценка 5».

Вычисление по шагам:

1. **Смотрим термы «оценка = 3»** в левой части таблицы:

$$\mu_{\text{Неуд.}}(3) = 0{,}2, \qquad \mu_{\text{Удовл.}}(3) = 0{,}5.$$

2. **Смотрим термы «посещаемость = 80 %»** в правой части таблицы:

$$\mu_{\text{Неуд}}(80\%) = 0, \qquad \mu_{\text{Удовл.}}(80\%) = 0{,}8, \qquad \mu_{\text{Хорошая}}(80\%) = 0{,}7.$$

3. **Находим в базе правил, какие правила дают на выходе «5».** По базе правил (см. выше) к итогу «5» ведёт только одно правило: «ЕСЛИ оценка Удовл. И посещаемость Хорошая, ТО итог = 5».

4. **Считаем силу этого правила** — конъюнкция «И» через min:

$$\min\big(\mu_{\text{Удовл.}}(3),\ \mu_{\text{Хорошая}}(80\%)\big) = \min(0{,}5;\ 0{,}7) = 0{,}5.$$

5. **Объединяем все правила, ведущие к «5»**, через max. Правило одно, поэтому итог не меняется:

$$F = \max(0{,}5) = 0{,}5.$$

**Ответ:** $F \approx 0{,}5$. То есть у человека с оценкой 3 и посещаемостью 80 % — умеренная (половинчатая) уверенность, что в зачётку пойдёт именно «5»: сдерживающий фактор здесь сама оценка ($\mu_{\text{Удовл.}}(3) = 0{,}5$ ниже, чем посещаемость).

**Высказывание 2:** «Человеку с оценкой 3 и посещаемостью 80 % будет поставлена (в зачётку) итоговая оценка 4».

Вычисление по шагам (входные термы те же, что и в высказывании 1):

1. **Термы «оценка = 3»:**

$$\mu_{\text{Неуд.}}(3) = 0{,}2, \qquad \mu_{\text{Удовл.}}(3) = 0{,}5.$$

2. **Термы «посещаемость = 80 %»:**

$$\mu_{\text{Неуд}}(80\%) = 0, \qquad \mu_{\text{Удовл.}}(80\%) = 0{,}8, \qquad \mu_{\text{Хорошая}}(80\%) = 0{,}7.$$

3. **Находим в базе правил, какие правила дают на выходе «4».** Их два:
   - «ЕСЛИ оценка Неуд. И посещаемость Хорошая, ТО итог = 4»;
   - «ЕСЛИ оценка Удовл. И посещаемость Удовл., ТО итог = 4».

4. **Считаем силу каждого правила** (конъюнкция «И» через min):

$$\min\big(\mu_{\text{Неуд.}}(3),\ \mu_{\text{Хорошая}}(80\%)\big) = \min(0{,}2;\ 0{,}7) = 0{,}2,$$

$$\min\big(\mu_{\text{Удовл.}}(3),\ \mu_{\text{Удовл.}}(80\%)\big) = \min(0{,}5;\ 0{,}8) = 0{,}5.$$

5. **Объединяем оба правила** через max:

$$F = \max(0{,}2;\ 0{,}5) = 0{,}5.$$

**Ответ:** $F \approx 0{,}5$ — та же степень уверенности, что и для итоговой «5» в высказывании 1. Совпадение не случайно: в обоих случаях «узким местом» (min) оказывается одна и та же величина — $\mu_{\text{Удовл.}}(3) = 0{,}5$, поскольку обе посещаемостные принадлежности ($0{,}8$ и $0{,}7$) её превышают. Значит, при такой оценке и посещаемости человек с равной уверенностью ($0{,}5$) может получить как «4», так и «5» — сама посещаемость их не различает, решающим является балл за оценку.

**Высказывание 3:** «Человеку с оценкой 3 и посещаемостью 80 % будет поставлена (в зачётку) итоговая оценка 3».

Вычисление по шагам (входные термы те же, что и в высказываниях 1–2):

1. **Термы «оценка = 3»:**

$$\mu_{\text{Неуд.}}(3) = 0{,}2, \qquad \mu_{\text{Удовл.}}(3) = 0{,}5.$$

2. **Термы «посещаемость = 80 %»:**

$$\mu_{\text{Неуд}}(80\%) = 0, \qquad \mu_{\text{Удовл.}}(80\%) = 0{,}8, \qquad \mu_{\text{Хорошая}}(80\%) = 0{,}7.$$

3. **Находим в базе правил, какие правила дают на выходе «3».** Их три:
   - «ЕСЛИ оценка Неуд. И посещаемость Неуд, ТО итог = 3»;
   - «ЕСЛИ оценка Неуд. И посещаемость Удовл., ТО итог = 3»;
   - «ЕСЛИ оценка Удовл. И посещаемость Неуд, ТО итог = 3».

4. **Считаем силу каждого правила** (конъюнкция «И» через min):

$$\min\big(\mu_{\text{Неуд.}}(3),\ \mu_{\text{Неуд}}(80\%)\big) = \min(0{,}2;\ 0) = 0,$$

$$\min\big(\mu_{\text{Неуд.}}(3),\ \mu_{\text{Удовл.}}(80\%)\big) = \min(0{,}2;\ 0{,}8) = 0{,}2,$$

$$\min\big(\mu_{\text{Удовл.}}(3),\ \mu_{\text{Неуд}}(80\%)\big) = \min(0{,}5;\ 0) = 0.$$

5. **Объединяем все три правила** через max:

$$F = \max(0;\ 0{,}2;\ 0) = 0{,}2.$$

**Ответ:** $F \approx 0{,}2$ — заметно ниже, чем для «4» и «5» ($0{,}5$ у обоих, см. высказывания 1–2). Причина: терм «Удовл.» оценки (его вес $0{,}5$) в правилах на «3» комбинируется только с посещаемостью «Неуд», а она при 80 % равна нулю — эта пара обнуляется. Остаётся только связка $\mu_{\text{Неуд.}}(3)=0{,}2$ с $\mu_{\text{Удовл.}}(80\%)=0{,}8$, которая и определяет итог.

**Свод по всем трём высказываниям** (оценка 3, посещаемость 80 %): $F(3) = 0{,}2$, $F(4) = 0{,}5$, $F(5) = 0{,}5$ — по совокупности данных наиболее вероятны «4» и «5», а не «3».

## Основные этапы построения механизма нечёткого вывода

Более подробная развёртка того же блока «нечёткий логический вывод», что на общей схеме выше (см. «Система нечёткого вывода»): там он был одним прямоугольником, здесь расписан по шагам:

```
Формирование базы правил
        ↓
Фаззификация входных переменных
        ↓
Агрегирование подусловий
        ↓
Активизация подзаключений
        ↓
Аккумулирование заключений
```

1. **Формирование базы правил** — сбор набора продукционных правил «ЕСЛИ ... ТО ...» (см. выше «Продукционное (нечёткое) правило»). Разовый, подготовительный этап: база правил не пересобирается при каждом новом входном значении.
2. **Фаззификация входных переменных** — то же самое, что фазификатор на общей схеме: перевод чёткого входного значения в степени принадлежности его термам (пример — шаг 1 в разборе с обогревателем: μ_холодно = 0,8, μ_тепло = 0,25).
3. **Агрегирование подусловий** — вычисление истинности составного антецедента, если он собран из нескольких условий связками «И» / «ИЛИ» (конъюнкция = min, дизъюнкция = max — операторы Заде; см. пример «оценка + посещаемость», где через min считалась сила каждого правила).
4. **Активизация подзаключений** — «усечение» терма-заключения (консеквента) каждого правила по силе его антецедента: например, в примере с обогревателем терм «сильная» отсекался на уровне 0,8, «средняя» — на уровне 0,25 (шаг 2 в том разборе).
5. **Аккумулирование заключений** — объединение всех усечённых заключений в одно итоговое нечёткое множество, обычно через max (в примере с обогревателем — объединение отсечённых термов «сильная» и «средняя» в одну область на графике).

**Примечание.** Дефазификация (превращение итогового нечёткого множества в одно чёткое число Y) в эту схему не входит — она идёт отдельным, следующим шагом уже за пределами блока «нечёткий вывод» (см. «Система нечёткого вывода» выше).

## Формирование базы правил

**База правил** систем нечёткого вывода формально представляет эмпирические знания или знания экспертов в конкретной предметной области. Используются правила нечётких продукций (см. выше «Продукционное (нечёткое) правило») — условия и заключения в них сформулированы в терминах нечётких лингвистических высказываний.

**База правил нечётких продукций** — конечное множество таких правил, согласованных относительно используемых в них лингвистических переменных. Чаще всего записывается в форме структурированного текста:

```
ПРАВИЛО 1: ЕСЛИ «Условие_1» ТО «Заключение_1» (F1)
ПРАВИЛО 2: ЕСЛИ «Условие_2» ТО «Заключение_2» (F2)
...
ПРАВИЛО n: ЕСЛИ «Условие_n» ТО «Заключение_n» (Fn)
```

Здесь **Fᵢ** (i ∈ 1, 2, …, n) — коэффициент определённости (весовой коэффициент) соответствующего правила, значение из интервала [0, 1]. По форме это сокращённая запись того же ядра A⇒B из общей схемы продукции «(i): Q;P;A⇒B;S,F,N» (см. выше) — только антецедент, консеквент и коэффициент F, без явных Q, P, S, N: на практике базу правил обычно и задают в этом сжатом виде.

**Важно не путать это Fᵢ с F из примера «оценка + посещаемость» выше.** Там F — это *вычисляемая* степень уверенности в конкретном высказывании («будет поставлена оценка 5») для конкретной пары входных значений, получаемая агрегированием (min) и аккумулированием (max) по всей базе правил. Здесь же Fᵢ — *статический* вес, заранее приписанный самому правилу при формировании базы (насколько это правило вообще заслуживает доверия), и он используется как один из множителей в дальнейших вычислениях, а не как их результат.

## Формирование базы правил: три множества

При задании (формировании) базы правил нечётких продукций нужно определить три множества:

- **множество правил** нечётких продукций (та самая база из предыдущего слайда — ПРАВИЛО_1, …, ПРАВИЛО_n):

$$P = \{R_1,\ R_2,\ \ldots,\ R_n\}$$

- **множество входных** лингвистических переменных:

$$V = \{\beta_1,\ \beta_2,\ \ldots,\ \beta_n\}$$

- **множество выходных** лингвистических переменных:

$$W = \{\omega_1,\ \omega_2,\ \ldots,\ \omega_s\}$$

Каждый элемент $R_i$ множества $P$ — это одно правило «ЕСЛИ Условие ТО Заключение (F)»: его антецедент собирается из термов переменных множества $V$ (входных, β — то же обозначение, что и в формальном определении лингвистической переменной выше), а консеквент — из термов переменных множества $W$ (выходных, обозначены уже не β, а ω — чтобы не путать со входными).

**Замечание про индексы.** На слайде и P, и V индексированы до $n$ — но это, скорее всего, независимые $n$ (число правил не обязано совпадать с числом входных переменных), а не намёк на то, что каждому правилу соответствует ровно одна входная переменная. Для выходных переменных использована уже отдельная буква $s$, что подтверждает: количества этих трёх множеств в общем случае разные.

## Фаззификация (формальное определение)

**Фаззификация** (введение нечёткости) — установление соответствия между численным (чётким) значением входной переменной системы нечёткого вывода и значением функции принадлежности соответствующего ей терма лингвистической переменной.

На этапе фаззификации значениям всех входных переменных, полученным извне (например, от датчиков), ставятся в соответствие конкретные значения функций принадлежности термов, которые используются в условиях (антецедентах) ядер правил базы нечётких продукционных правил (см. выше «Формирование базы правил»).

**Формальный критерий завершённости.** Фаззификация считается выполненной, если найдены степени истинности $\mu_A(x)$ всех элементарных логических высказываний вида «β ЕСТЬ ά» (та самая запись «β ЕСТЬ α», что и в «Вывод правила. Импликация „ЕСЛИ..., ТО...“» выше), входящих в антецеденты нечётких продукционных правил — где ά — некоторый терм с известной функцией принадлежности $\mu_A(x)$, а x — чёткое численное значение, принадлежащее универсуму лингвистической переменной β.

**Связь с уже разобранным примером.** Именно это и происходило на шаге 1 в разборе с обогревателем (см. «Пример: полный цикл нечёткого вывода»): β = «температура», чёткое значение x = 12°C, термы α ∈ {холодно, тепло, горячо}; фаззификация нашла $\mu_{\text{холодно}}(12)=0{,}8$, $\mu_{\text{тепло}}(12)=0{,}25$, $\mu_{\text{горячо}}(12)=0$ — то есть степени истинности трёх высказываний «температура ЕСТЬ холодно», «температура ЕСТЬ тепло», «температура ЕСТЬ горячо».

**Про написание термина.** На этом слайде и далее в источнике используется написание «фаззификация» (двойное «з»); в разборе с обогревателем выше по тексту конспекта использовался вариант «фазификация» (одно «з») — это то же самое понятие, орфографический разнобой источника, не два разных термина.

## Фаззификация (2): процедура и критерий по всей базе правил

Уточнение к определению фаззификации с предыдущего слайда:

- Фаззификация — это не просто «этап» в общей схеме вывода, а сама **процедура** (процесс) нахождения значений функций принадлежности нечётких множеств (термов) по обычным (не нечётким) исходным данным. Отсюда и второе название — «введение нечёткости».
- **Цель** этапа — установить соответствие между конкретным (численным) значением ОДНОЙ входной переменной и значением функции принадлежности соответствующего ей терма входной лингвистической переменной.

**Уточнение критерия завершённости.** На предыдущем слайде критерий был сформулирован через антецеденты правил вообще; здесь появляется термин **«подусловие»** — то самое слово, что уже встречалось в списке этапов построения механизма нечёткого вывода («Агрегирование подусловий», см. выше «Основные этапы построения механизма нечёткого вывода»). Подусловие — это, по сути, то же элементарное высказывание «β ЕСТЬ α» (см. предыдущий слайд), но явно рассматриваемое как составная часть (возможно, сложного, из нескольких условий через «И»/«ИЛИ») антецедента одного правила.

Фаззификация считается законченной, когда найдены значения функций принадлежности **для каждого подусловия всех правил всей базы** — не одного правила и не одной входной переменной, а по базе правил целиком. Именно этот набор значений (μ по всем подусловиям) и передаётся дальше на этап агрегирования подусловий.

## Агрегирование

**Агрегирование** — процедура определения степени истинности условий по каждому из правил системы нечёткого вывода. Это тот же этап «Агрегирование подусловий» из общей схемы построения механизма нечёткого вывода (см. выше «Основные этапы построения механизма нечёткого вывода») — здесь он показан на конкретных входных данных.

Таблица — значения функции принадлежности для терма **«Низкий»** переменной «Рост» (тот самый терм, что войдёт в антецедент одного из правил на этапе агрегирования):

| Рост | 160 | 165 | 170 | 175 | 180 |
|---|---|---|---|---|---|
| μ_Низкий | 1 | 0,8 | 0,6 | 0,4 | 0,2 |

Зависимость линейная: $\mu_{\text{Низкий}}(x) = 1 - 0{,}04\cdot(x-160)$ (шаг 5 см роста ↔ шаг 0,2 принадлежности).

**Судя по всему, это начало нового развёрнутого примера** (переменная «Рост», термы вроде «Низкий» — перекликается с более ранним разбором роста в примере S-функции, см. «Конкретный пример: терм «высокий» по росту»), который, вероятно, продолжится на следующих слайдах (активизация, аккумулирование, дефазификация) — пока видна только эта таблица, дальнейшие расчёты по ней не приведены.

## Активизация

**Активизация** в системах нечёткого вывода — процедура (процесс) нахождения степени истинности каждого из элементарных логических высказываний (**подзаключений**), составляющих консеквенты ядер всех нечётких продукционных правил.

Поскольку заключения делаются относительно ВЫХОДНЫХ лингвистических переменных, степеням истинности элементарных подзаключений при активизации ставятся в соответствие элементарные функции принадлежности (термов выходных переменных).

**Структурная параллель с фаззификацией.** Активизация устроена симметрично фаззификации (см. выше), но работает с другой стороной правила:

| | Фаззификация | Активизация |
|---|---|---|
| Работает с | подусловиями (антецедент) | подзаключениями (консеквент) |
| Переменные | входные | выходные |
| Источник значений μ | внешние данные (например, датчики) | сила сработавшего правила (результат агрегирования) |

«Подзаключение» — тот же термин из списка этапов выше («Активизация подзаключений», см. «Основные этапы построения механизма нечёткого вывода»), и по форме то же элементарное высказывание «β ЕСТЬ α», что и подусловие при фаззификации, — только здесь β уже выходная переменная.

**Связь с уже разобранным примером.** Именно это происходило на шаге 2 в разборе с обогревателем (см. «Пример: полный цикл нечёткого вывода»): терм-заключение «сильная» получил (был активизирован) степенью истинности 0,8 — по силе сработавшего правила «холодно → сильная», найденной на этапе агрегирования; терм «средняя» — степенью 0,25. Графически это и есть «отсечение по высоте» термов вывода на том рисунке.

## Аккумулирование

**Аккумуляция** (аккумулирование) в системах нечёткого вывода — процесс нахождения функции принадлежности для **каждой из выходных лингвистических переменных** (в отличие от активизации, которая находит истинность отдельных подзаключений по одному).

- Цель — объединить все степени истинности подзаключений (уже найденные на этапе активизации) в одну функцию принадлежности каждой выходной переменной.
- Формально: результат аккумуляции для выходной переменной — это объединение нечётких множеств всех её подзаключений из базы правил.
- Объединение считается классически, как обычное объединение нечётких множеств (см. выше «Операции над нечёткими множествами») — через max:

$$\forall x \in X:\ \mu_{A\cup B}(x) = \max\{\mu_A(x),\ \mu_B(x)\}$$

**Связь с примером обогревателя.** Это и есть шаг «отсечённые термы объединяются (max) в одно результирующее нечёткое множество» из разбора с обогревателем (см. «Пример: полный цикл нечёткого вывода»): термы «сильная» (отсечена активизацией на 0,8) и «средняя» (отсечена на 0,25) аккумулированием объединялись через max в одно итоговое нечёткое множество мощности, которое затем дефазифицировалось в Y ≈ 72,9%.

**Активизация vs аккумулирование** (обе работают с подзаключениями, легко спутать):
- **активизация** — присваивает степень истинности КАЖДОМУ отдельному подзаключению по отдельности (одно на правило);
- **аккумулирование** — ОБЪЕДИНЯЕТ (через max) уже готовые подзаключения в одну общую функцию принадлежности, причём отдельно для каждой выходной переменной (если их несколько — у каждой своё аккумулированное множество).

## Дефаззификация (формальное определение)

**Дефаззификация** в системах нечёткого вывода — процесс перехода от функции принадлежности выходной лингвистической переменной к её чёткому (числовому) значению.

**Цель** — используя результаты аккумуляции по каждой выходной лингвистической переменной, получить количественное значение для неё, которое используется внешними по отношению к системе нечёткого вывода устройствами (например, исполнительными механизмами интеллектуальной САУ — системы автоматического управления).

**Подтверждение более ранней заметки.** В ячейке «Основные этапы построения механизма нечёткого вывода» уже отмечалось, что дефазификация не входит в диаграмму из пяти этапов (формирование базы правил → фаззификация → агрегирование → активизация → аккумулирование), а идёт отдельным, следующим шагом. Этот слайд подтверждает это явно: дефаззификация здесь описана уже после аккумулирования, отдельным блоком.

**Связь с уже разобранным примером.** Это в точности шаг 3 в разборе с обогревателем (см. «Пример: полный цикл нечёткого вывода»): аккумулированное нечёткое множество мощности превращалось в одно число Y ≈ 72,9%. Там был использован упрощённый способ (взвешенное усреднение представительных значений термов) — конкретные общепринятые методы дефаззификации (центр тяжести и др.), судя по всему, разбираются на следующих слайдах.

**Про написание термина.** Как и с фаззификацией/фазификацией выше, источник снова колеблется: здесь — «дефаззификация» (двойное «з»), а в общей схеме нечёткого вывода и в разборе с обогревателем использовался вариант «дефазификация» (одно «з») — то же самое понятие, орфографический разнобой источника.

## Правила дефаззификации (по содержанию — методы)

**Заголовок слайда неточен.** Слайд называется «Правила дефаззификации», но по содержанию перечислены не правила, а конкретные **методы** — способы вычисления итогового чёткого значения y по накопленной (аккумулированной) функции принадлежности μ(x). Похоже на неточность формулировки источника — как и в паре других мест этого конспекта, где заголовок/термин не вполне соответствует содержимому.

Перечислены три метода:

1. **Метод левого модального значения:** $y = x_{\min}$ — берётся крайняя левая точка модального множества (там, где μ(x) впервые достигает максимума).
2. **Метод правого модального значения:** $y = x_{\max}$ — симметрично, крайняя правая точка максимума.
3. **Метод центра площади** (биссектриса площади): вычисляется абсцисса y, делящая пополам площадь, ограниченную кривой функции принадлежности μ(x).

**Уточнение к более ранней заметке.** В ячейке «Дефаззификация (формальное определение)» была высказана догадка, что дальше разбирается метод «центр тяжести» — на деле это **метод центра площади / биссектрисы площади**, который *не то же самое*, что центр тяжести (центроид): центроид берёт взвешенное среднее (первый момент) площади под кривой, а биссектриса площади — точку, слева и справа от которой площади равны. В общем случае эти два y не совпадают, хотя оба относятся к семейству «центроидных» методов.

На графике справа — иллюстрация из нескольких перекрывающихся кусочно-линейных функций принадлежности с отмеченными точками по обеим осям (0,08–0,41 по x; 0,16–0,83 по y); по одному скриншоту однозначно сопоставить конкретные кривые с тремя перечисленными методами не получается.

## Применение нормализации

**Нормализация** — приведение разномасштабных величин к единой шкале, например к интервалу [0; 1] или [0; 100].

Примеры из слайда:

- **Нормализованная оценка** (исходная шкала 1–5): 0,2; 0,4; 0,6; 0,8; 1 — то есть оценка/5.
- **Нормализованная посещаемость** (исходная шкала 0–100%): 0; 0,1; …; 1 — то есть посещаемость/100.

**Зачем это здесь.** Термы разных лингвистических переменных обычно заданы на разных по масштабу носителях (возраст 0–120 лет, посещаемость 0–100%, оценка 1–5 баллов и т.п.) — нормализация приводит их к общему [0,1], что упрощает единообразную запись функций принадлежности и сравнение термов между собой (то же самое неявно уже делалось в разборе S-функции для роста, где рост тоже нормировался: $u = (\text{рост}-150)/50$, см. «Конкретный пример: терм «высокий» по росту»).

На слайде показан график двух треугольных термов на нормализованной шкале [0,1] (по оси x — метки 0,1…1), пересекающихся примерно на уровне x≈0,7: левый терм с пиком около x=0,6, правый — с плато на максимуме начиная примерно с x=0,8.

## Варианты построения μ-функций

Пример показывает, что одну и ту же составную нечёткую переменную B можно получить из нескольких исходных переменных — здесь «Рост» и «Техника» (уже встречались раньше в конспекте) и «Рез» (по всей видимости, «Результат»; отдельного определения на этом слайде нет) — по-разному, комбинируя их через операции пересечения/объединения (см. «Операции над нечёткими множествами» выше: ∩ = min, ∪ = max):

- $\mu(B_1) = \mu(\text{Рост}) \cap \mu(\text{Техника}) \cap \mu(\text{Рез})$ — полная конъюнкция («И» по всем трём), самый строгий вариант: B1 «сработает» только если все три условия выполнены хорошо одновременно.
- $\mu(B_2) = \mu(\text{Рост}) \cup \mu(\text{Техника}) \cup \mu(\text{Рез})$ — полная дизъюнкция («ИЛИ» по всем трём), самый мягкий вариант: достаточно, чтобы хорошо выполнялось хотя бы одно условие.
- $\mu(B_3) = \big(\mu(\text{Рост}) \cap \mu(\text{Техника})\big) \cup \mu(\text{Рез})$ — смешанный: (Рост И Техника) ИЛИ Рез.
- $\mu(B_4) = \mu(\text{Рост}) \cup \big(\mu(\text{Техника}) \cap \mu(\text{Рез})\big)$ — смешанный: Рост ИЛИ (Техника И Рез).

**Замечание про запись B1.** На слайде формула для B1 дана избыточно — «min(μ(Рост)) ∩ min(μ(Техника)) ∩ min(μ(Рез))» — хотя ∩ само по себе уже есть min (см. выше), лишний min(...) вокруг каждого отдельного терма ничего не меняет, просто дублирует обозначение.

**Неоднозначность в B3 и B4.** Без скобок порядок выполнения ∩ и ∪ в одной строке формально не определён: B3, например, можно прочитать и как (Рост ∩ Техника) ∪ Рез, и как Рост ∩ (Техника ∪ Рез) — результат разный. Выше я расставил скобки по аналогии с B1/B2 (где операции однородны) и по обычному приоритету «∩ старше ∪» (как «И» старше «ИЛИ» в булевой логике) — но явных скобок в источнике нет, это моя интерпретация, а не то, что написано на слайде.

## Пример: комната, обогреватель и кондиционер (два входа, полный цикл)

Собственный пример (не со слайдов) — чтобы пройти весь цикл нечёткого вывода на знакомых переменных, но с **двумя входами и двумя выходами сразу**. В отличие от более ранней версии этого примера, здесь антецеденты составные (по два подусловия на правило), поэтому агрегирование не сводится к копированию значения фаззификации, а реально считает min по двум подусловиям — как в примере «оценка + посещаемость» выше, только на новых переменных.

**Переменные:**
- Вход 1: X₁ = «температура» (°C) — термы холодно/тепло/горячо, те же формулы μ, что и в разборе «Пример: полный цикл нечёткого вывода» выше.
- Вход 2: X₂ = «влажность» (%) — термы сухо/нормально/влажно, формулы μ:

$$\mu_{\text{сухо}}(h) = \begin{cases} 1, & h \le 30 \\ \dfrac{50-h}{20}, & 30 < h \le 50 \\ 0, & h > 50 \end{cases} \qquad \mu_{\text{нормально}}(h) = \begin{cases} 0, & h \le 30 \\ \dfrac{h-30}{20}, & 30 < h \le 50 \\ 1, & 50 < h \le 60 \\ \dfrac{80-h}{20}, & 60 < h \le 80 \\ 0, & h > 80 \end{cases} \qquad \mu_{\text{влажно}}(h) = \begin{cases} 0, & h \le 60 \\ \dfrac{h-60}{20}, & 60 < h \le 80 \\ 1, & h > 80 \end{cases}$$

- Выход 1: «мощность обогревателя», термы слабая/средняя/сильная (представительные значения 20/50/80%).
- Выход 2: «мощность кондиционера», термы выключен/слабо/сильно (представительные значения 0/40/90%).

**База правил** (составной антецедент — «И» по обоим входам; составной консеквент — по обоим выходам):

1. ЕСЛИ температура ЕСТЬ холодно И влажность ЕСТЬ сухо ТО (обогреватель ЕСТЬ сильная) И (кондиционер ЕСТЬ выключен)
2. ЕСЛИ температура ЕСТЬ тепло И влажность ЕСТЬ нормально ТО (обогреватель ЕСТЬ средняя) И (кондиционер ЕСТЬ слабо)
3. ЕСЛИ температура ЕСТЬ горячо И влажность ЕСТЬ влажно ТО (обогреватель ЕСТЬ слабая) И (кондиционер ЕСТЬ сильно)

Возьмём X₁ = 27°C, X₂ = 65%.

**1. Фаззификация.**

Температура (по формулам из разбора с обогревателем):
$$\mu_{\text{холодно}}(27)=0,\qquad \mu_{\text{тепло}}(27)=\frac{30-27}{8}=0{,}375,\qquad \mu_{\text{горячо}}(27)=\frac{27-22}{8}=0{,}625.$$

Влажность:
$$\mu_{\text{сухо}}(65)=0,\qquad \mu_{\text{нормально}}(65)=\frac{80-65}{20}=0{,}75,\qquad \mu_{\text{влажно}}(65)=\frac{65-60}{20}=0{,}25.$$

**2. Агрегирование.** Теперь у каждого правила по два подусловия, соединённых «И» — сила правила = min от них:

- правило 1: $\min(\mu_{\text{холодно}},\ \mu_{\text{сухо}}) = \min(0;\ 0) = 0$
- правило 2: $\min(\mu_{\text{тепло}},\ \mu_{\text{нормально}}) = \min(0{,}375;\ 0{,}75) = 0{,}375$
- правило 3: $\min(\mu_{\text{горячо}},\ \mu_{\text{влажно}}) = \min(0{,}625;\ 0{,}25) = 0{,}25$

Обратите внимание на правило 3: по одной температуре его сила была бы 0,625, но влажность оказалась «слабым звеном» (min выбирает меньшее из двух подусловий) и утянула итоговую силу правила вниз до 0,25 — это и есть агрегирование в деле, в отличие от предыдущей версии примера, где сила правила просто повторяла фаззификацию.

**3. Активизация.** Сила правила применяется к обоим его заключениям (они относятся к разным выходным переменным, поэтому дополнительно объединять их между собой не нужно):

- правило 1 (сила 0): обогреватель.сильная → 0, кондиционер.выключен → 0;
- правило 2 (сила 0,375): обогреватель.средняя → 0,375, кондиционер.слабо → 0,375;
- правило 3 (сила 0,25): обогреватель.слабая → 0,25, кондиционер.сильно → 0,25.

**4. Аккумулирование** (отдельно для каждого выхода, объединение через max):

- обогреватель: {средняя: 0,375; слабая: 0,25};
- кондиционер: {слабо: 0,375; сильно: 0,25}.

![Накопленный результат для «обогреватель»: усечённые термы слабая (0,25) и средняя (0,375), объединённые в одну фигуру; пунктиром — исходные термы и итоговая точка дефаззификации Y≈38%](images/heater_accumulation_example.svg)

На картинке — расшифровка строки выше в виде графика (для выхода «обогреватель»): серым пунктиром — исходные полные термы слабая/средняя/сильная — для сравнения, показывают, какими они были бы «в полный рост». Синяя заливка — то, что получилось после аккумулирования: слева треугольник «слабая», усечённый по высоте 0,25, справа — «средняя», усечённая по высоте 0,375, объединённые через max (там, где они перекрываются между x=27,5 и x=31,25, побеждает более высокая — средняя). Красная точка — куда упадёт дефаззификация (Y≈38%): это не просто «середина» фигуры на глаз, а именно взвешенное среднее представительных точек с весами 0,25 и 0,375, как посчитано ниже.

**5. Дефаззификация** (упрощённо, взвешенное среднее представительных значений):

$$Y_{\text{обогреватель}} = \frac{0{,}25\cdot20 + 0{,}375\cdot50}{0{,}25+0{,}375} = \frac{5+18{,}75}{0{,}625} = \frac{23{,}75}{0{,}625} = 38\%$$

$$Y_{\text{кондиционер}} = \frac{0{,}375\cdot40 + 0{,}25\cdot90}{0{,}375+0{,}25} = \frac{15+22{,}5}{0{,}625} = \frac{37{,}5}{0{,}625} = 60\%$$

**Итог.** При 27°C и влажности 65% обогреватель выходит на ≈38%, кондиционер — на ≈60%: тёплая и довольно влажная комната скорее просит охлаждения и осушения, чем обогрева — результат разумный.

**Отличие от предыдущей версии примера.** Здесь сумма весов в знаменателе (0,625) уже **не равна 1** — в отличие от предыдущей версии, где это было совпадением из-за формул температуры на [22, 30]. Теперь видно, зачем в формуле взвешенного среднего вообще нужно делить на сумму весов, а не только умножать и складывать: веса, полученные агрегированием по нескольким подусловиям, в общем случае не образуют «удобную» единицу.
