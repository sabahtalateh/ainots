## Определение графа

**Графом** $G(V, E)$ называется пара множеств, где:

- $V$ — совокупность **объектов** (вершин, vertices);
- $E$ — совокупность **пар отношений** между этими объектами (рёбер, edges).

![Пример графа](images/graph_def.svg)

Вершины изображают кружками, отношения между ними — линиями (рёбрами). У рёбер может быть **направление** (стрелка) и **вес** (число рядом с ребром — например, стоимость, расстояние, пропускная способность).

Графом можно описать самые разные структуры: транспортные и социальные сети, деревья (иерархии), схемы процессов и т. д.

## Что можно описать с помощью графа

Граф — универсальный язык для любых структур «объекты + связи». Примеры:

- **Транспортные сети** (схема метро): вершины — станции, рёбра — перегоны; задача — кратчайший маршрут, пассажиропотоки.
- **Карты и территории** (регионы страны): вершины — области, рёбра — общая граница; задача — раскраска карты, соседство.
- **Схемы процессов** (например, работа тормозной системы): вершины — узлы/детали, рёбра — «передаёт усилие», «прижимает»; описывают причинно-следственные цепочки.
- **Социальные сети**: вершины — люди, рёбра — знакомства/связи; задачи — сообщества, влиятельные узлы, связность.

## ИИ и графы

Графы лежат в основе многих методов искусственного интеллекта:

- **Языковые модели (LM)** — особая разновидность графов.
- **Нейросеть** — тоже особый, сильно упрощённый подвид графовых моделей (узлы-нейроны, связи-веса).
- **ИИ строит социальные графы** — выявляет связи между людьми, сообщества.
- **ИИ строит графы знаний** (knowledge graphs) — связывает факты и понятия в сеть «сущность → отношение → сущность».

То есть граф — это и структура данных для ИИ, и способ представления самой модели.

## Ориентированные (орграфы) и неориентированные графы

**Ориентированный граф** (кратко **орграф**) — граф, рёбрам которого присвоено **направление**. Направленные рёбра называют также **дугами**.

- В **орграфе** ребро $A \to B$ означает связь «от $A$ к $B$» (в одну сторону). Пример: подписки, ссылки, поток, последовательность действий.
- В **неориентированном** графе ребро $A - B$ означает **взаимную** связь (в обе стороны). Пример: дружба, общая граница, соединение проводом.

![Орграф и неориентированный граф](images/directed_undirected.svg)

Ребро, соединяющее вершину саму с собой (как петля у вершины 1 справа), называют **петлёй**.

## Смежные вершины

- Вершины графа называются **смежными**, если они соединены хотя бы одним ребром.
- **Степень вершины** — количество связанных (**инцидентных**) с ней рёбер. Для орграфа отдельно считают исходящие (полустепень исхода) и входящие (полустепень захода).

![Смежные вершины](images/adjacent.svg)

На рисунке, например, $v_1$ смежна с $v_2, v_3, v_4, v_5$ — значит её **степень равна 4**. Остальные вершины ($v_2, v_3, v_4, v_5$) имеют степень 3.

**Инцидентность** — это отношение между вершиной и ребром: ребро инцидентно двум своим концевым вершинам, а вершина инцидентна всем рёбрам, которые из неё выходят.

## Степень вершины

- **Степень** (или **валентность**) вершины графа $G$ — количество рёбер, **инцидентных** вершине $v_i$ (для орграфа берут входящие рёбра).
- При подсчёте степени неориентированное ребро-**петля** учитывается **дважды** (оба конца петли — эта же вершина).
- Вершина степени **0** — **изолированная** (ни с чем не соединена).
- Вершина степени **1** — **концевая, висячая** или **лист графа**. Ребро, инцидентное такой вершине, — **висячее** (англ. terminal / pendant edge, end-edge).

![Степени вершин](images/degree.svg)

На рисунке каждая вершина подписана **своей степенью**:

- «**0**» слева — изолированная;
- «**1**» — концевая (лист), её ребро висячее;
- «**5**» — три обычных ребра плюс **петля** (даёт +2): $3 + 2 = 5$.

## Матрица смежности для неориентированного графа

**Матрица смежности** — квадратная таблица размером $n \times n$ (по числу вершин). В клетке на пересечении строки $i$ и столбца $j$ стоит:

- **1**, если вершины $i$ и $j$ **смежны** (соединены ребром);
- **0**, если ребра между ними нет.

Для приведённого графа (все пять вершин соединены каждая с каждой — полный граф):

| | Е | ВИ | Н | О | ВВ |
|---|---|---|---|---|---|
| **Е** | 0 | 1 | 1 | 1 | 1 |
| **ВИ** | 1 | 0 | 1 | 1 | 1 |
| **Н** | 1 | 1 | 0 | 1 | 1 |
| **О** | 1 | 1 | 1 | 0 | 1 |
| **ВВ** | 1 | 1 | 1 | 1 | 0 |

![Полный граф](images/k5.svg)

Свойства матрицы смежности **неориентированного** графа:

- она **симметрична** относительно главной диагонали ($a_{ij} = a_{ji}$), ведь связь взаимна — поэтому на слайде заполняют только верхний треугольник;
- на **главной диагонали** нули (если нет петель);
- **сумма единиц в строке** = степень соответствующей вершины (здесь у каждой вершины степень 4).

## Представление графа как матрицы

Если рёбра имеют **веса** (и направление), в клетки матрицы записывают не 0/1, а **вес ребра**. Пустая клетка (или `NULL`) — ребра нет.

Для нашего взвешенного орграфа (строка — откуда, столбец — куда):

| из \ в | A | B | C | D | E | F |
|---|---|---|---|---|---|---|
| **A** | 0 | 4 | 2 | — | — | — |
| **B** | — | 0 | 5 | 10 | — | — |
| **C** | — | — | 0 | — | 3 | — |
| **D** | — | — | — | 0 | — | 11 |
| **E** | — | — | — | 4 | 0 | — |
| **F** | — | — | — | — | — | 0 |

![Взвешенный орграф](images/graph_def.svg)

Здесь, например, ячейка (A, B) = 4 — это дуга $A \to B$ весом 4. В отличие от неориентированного случая, матрица орграфа **несимметрична**: ячейка (B, A) пуста (`NULL`), потому что обратной дуги $B \to A$ нет.

## Мультиграфы и псевдографы

- **Мультиграф** — граф, у которого между вершинами может быть **более одной связи** (**кратные рёбра**). Пример: два города, соединённые несколькими дорогами.
- **Псевдограф** — если в мультиграфе присутствуют ещё и **петли** (ребро из вершины в саму себя), такой граф называют псевдографом.

![Мультиграф и псевдограф](images/multigraph.svg)

Иерархия: обычный граф (не более одного ребра между парой, без петель) ⊂ **мультиграф** (кратные рёбра) ⊂ **псевдограф** (кратные рёбра + петли).

## Изоморфные графы

Графы $G_1$ и $G_2$ **изоморфны**, если существует **взаимно-однозначное соответствие** (биекция) между их множествами вершин, **сохраняющее отношения смежности**: если две вершины соединены в $G_1$, то соответствующие им вершины соединены и в $G_2$ (и наоборот).

Проще говоря — это **один и тот же граф**, просто **нарисованный по-разному** (вершины переставлены, рёбра изогнуты). Структура связей одинаковая.

![Изоморфные графы](images/isomorphic.svg)

Оба рисунка — один граф на 5 вершинах с одним и тем же набором связей; отличается только расположение вершин на плоскости.

**Пример из практики:** социальные графы, построенные из разных источников (разные соцсети), могут описывать одни и те же связи между людьми — то есть быть изоморфными.

Изоморфные графы имеют одинаковые: число вершин, число рёбер, набор степеней вершин. Но совпадение этих признаков ещё не гарантирует изоморфизм — нужно именно соответствие, сохраняющее смежность.

## Подграф

Граф $G'(V', U')$ называется **подграфом** графа $G(V, U)$, если:

- подмножество вершин $V' \subseteq V$ и подмножество рёбер $U' \subseteq U$;
- любое ребро $\{v_i, v_j\}$ из $U'$ соединяет только вершины, принадлежащие $V'$ (нельзя взять ребро, у которого один конец вне $V'$).

То есть подграф — это «часть» графа: берём **некоторые** вершины и **некоторые** рёбра между ними.

**Порождённый (индуцированный) подграф** — если в него включены **все** рёбра исходного графа, оба конца которых попали в выбранное множество вершин $V'$. Иначе говоря: выбираем вершины $V'$ — и автоматически берём все рёбра между ними.

![Граф для подграфа](images/subgraph_base.svg)

Например, если из этого графа взять вершины $\{v_1, v_2, v_3\}$ и все рёбра между ними — получим **индуцированный** подграф на этих трёх вершинах.

## Подграф. Примеры

Возьмём из исходного графа (слева) три вершины $\{v_1, v_2, v_3\}$. В исходном графе между ними есть все три ребра: $v_1\!-\!v_2$, $v_2\!-\!v_3$, $v_1\!-\!v_3$.

![Исходный граф](images/subgraph_base.svg)

![Индуцированный и неиндуцированный подграфы](images/subgraph_examples.svg)

- **Индуцированный (порождённый) подграф** на $\{v_1,v_2,v_3\}$ — берём **все** рёбра между этими вершинами → получается треугольник.
- **Неиндуцированный подграф** на тех же вершинах — берём **не все** рёбра (например, пропустили $v_2\!-\!v_3$) → получается путь $v_2 - v_1 - v_3$. Это тоже подграф, но не индуцированный, потому что в исходном графе ребро $v_2\!-\!v_3$ было, а мы его не взяли.

Разница: индуцированный подграф однозначно задаётся **выбором вершин** (рёбра берутся автоматически), а неиндуцированный — ещё и произвольным выбором рёбер.

## Понятие дерева

**Дерево** — это:

- **связный ацикличный граф** (нет циклов, и из любой вершины можно добраться в любую другую), или, что то же самое,
- граф, в котором между **любыми двумя вершинами** существует **единственная простая цепь** (ровно один путь без повторений).

У дерева из $n$ вершин всегда ровно $n-1$ рёбер. Есть **корень** (верхняя вершина), **листья** (концевые вершины), **родители** и **потомки**.

```mermaid
graph TD
  1((1)) --> 2((2))
  1 --> 3((3))
  1 --> 4((4))
  2 --> 5((5))
  2 --> 6((6))
  5 --> 9((9))
  5 --> 10((10))
  4 --> 7((7))
  4 --> 8((8))
  7 --> 11((11))
  7 --> 12((12))
```

Здесь вершина **1** — корень; **9, 10, 6, 3, 11, 12, 8** — листья. Между любыми двумя вершинами — ровно один путь.

## Задача о кёнигсбергских мостах

**Можно ли посетить все части города, не проходя дважды по одному и тому же мосту?**

Кёнигсберг стоял на реке Прегель, которая делила город на **четыре части**: верхний берег, нижний берег и два острова (Большой и Малый). Части соединяли **семь мостов**.

**Основная проблема** заключалась в поиске маршрута прогулки, который позволял бы пересечь каждый из семи мостов **ровно один раз**. Местные жители долго пытались найти такой путь, но безуспешно.

![Кёнигсбергские мосты](images/konigsberg.svg)

Задачу решил **Леонард Эйлер** (1736): он свёл карту к графу и доказал, что такого маршрута не существует. Эта работа считается **началом теории графов**.

## Граф кёнигсбергских мостов

Эйлер свёл карту к графу: **вершины** — четыре части суши (два берега и два острова), **рёбра** — мосты между ними.

Это **мультиграф**, т. к. Большой остров соединён с берегами **двумя** мостами каждый, и по ним можно двигаться в любом направлении (граф неориентированный).

![Граф кёнигсбергских мостов](images/konigsberg_graph.svg)

Степени вершин:

- **Большой остров** — 5 (2 к верхнему + 2 к нижнему + 1 к Малому);
- **Малый остров** — 3 (по одному к Большому, верхнему и нижнему);
- **Верхний берег** — 3;
- **Нижний берег** — 3.

**Все четыре вершины имеют нечётную степень.** Именно это и делает задачу неразрешимой (см. следующий слайд).

## Решение Эйлера

Речь о том, можно ли обойти граф **одним росчерком** — начертить его, не отрывая карандаша от бумаги и не проводя ни одно ребро дважды (**эйлеров путь**).

- **Число нечётных вершин** (вершин, к которым ведёт нечётное число рёбер) графа всегда **чётно**. Не может существовать граф с нечётным числом нечётных вершин.
- Если **все вершины чётные** — граф можно начертить без отрыва карандаша, начав с **любой** вершины и завершив в **той же** (замкнутый эйлеров цикл).
- Если **ровно две вершины нечётные** — граф можно начертить одним росчерком, но начинать нужно **в одной** нечётной вершине, а закончить **в другой**.
- Если нечётных вершин **больше двух** — такой обход **невозможен**.

**Итог для Кёнигсберга:** у графа мостов **все четыре** вершины нечётные (больше двух) → пройти по каждому мосту ровно один раз **нельзя**. Это и доказал Эйлер.

## Почему это работает (смысл)

Идея — в подсчёте «входов и выходов» каждой вершины.

Представим, что мы гуляем и идём по каждому мосту **ровно один раз**. Возьмём любую вершину (часть суши), которая **не** старт и **не** финиш маршрута. Каждый раз, когда мы туда **заходим** по одному мосту, мы обязаны оттуда **выйти** по другому. Зашёл — вышел: мосты такой вершины расходуются **парами**.

Значит, у любой **транзитной** (промежуточной) вершины число мостов должно быть **чётным** — иначе останется «лишний» мост, по которому мы либо зайдём и застрянем, либо вообще не сможем зайти.

Нечётное число мостов допустимо только у **двух** особых точек:

- вершина **старта** — первый мост даёт «выход» без парного входа;
- вершина **финиша** — последний мост даёт «вход» без парного выхода.

Отсюда правило про эйлеров путь:

- **0 нечётных вершин** → маршрут есть и он замкнутый (старт = финиш);
- **ровно 2 нечётные** → маршрут есть: старт в одной, финиш в другой;
- **больше 2 нечётных** → маршрута нет.

**Кёнигсберг:** все четыре части суши имеют нечётное число мостов (5, 3, 3, 3). Четыре нечётные вершины — это больше двух, а «потратить» нечётность можно максимум в двух точках (старт и финиш). Две лишние нечётные вершины пристроить некуда → пройти по всем семи мостам ровно один раз **невозможно**.

**Аналогия:** конверт (домик с крестом) рисуется одним росчерком, потому что у него ровно две нечётные вершины. У графа мостов их четыре — поэтому не выходит.

## Граф Интернета (веб-граф)

Интернет удобно представить графом:

- **вершины** (множество $V$) — это **страницы** в сети;
- **рёбра** (множество $U$) — это **ссылки** (гиперссылки) между страницами.

Свойства веб-графа:

- **Орграф** — ссылка направлена: страница $A$ ссылается на $B$, но $B$ не обязана ссылаться на $A$.
- Может быть **мультиграфом** — с одной страницы на другую бывает несколько ссылок (кратные рёбра).
- Может содержать **петли** (страница ссылается сама на себя) → то есть быть **псевдографом**.

На веб-графе работают алгоритмы ранжирования (например, **PageRank**): важность страницы тем выше, чем больше на неё ведёт ссылок с других важных страниц.

## Социальный граф

**Социальный граф** — граф, узлы которого представлены **социальными объектами**: пользовательские профили с атрибутами (имя, день рождения, родной город), сообщества, медиаконтент и т. д., а **рёбра** — социальными связями между ними.

**Неявный социальный граф** — граф, который можно **вывести (вычислить)** на основе взаимодействий пользователя со своими «друзьями» и группами «друзей» в соцсети. В отличие от обычного социального графа, здесь **нет явного указания «друзей»** — то есть нет явно заданных социальных связей; связи восстанавливаются по поведению (переписка, лайки, совместные события).

Дополнительно: [sociology_encyclopedy.academic.ru — «Анализ социальных сетей»](https://sociology_encyclopedy.academic.ru/47/)

## Социальный граф сообществ

Содержит объекты **разного уровня**, например:

- **отдельных людей**;
- **группы** (сообщества).

![Социальный граф сообществ](images/community_graph.svg)

Внутри сообщества связи **плотные** (люди активно взаимодействуют друг с другом), между сообществами — **редкие** (единичные связи через «мостиковых» участников). Именно по такой плотности алгоритмы выделяют **кластеры-сообщества** в графе.

## Пример кластеризованного социального графа

Граф разбит на **кластеры** (сообщества) — группы вершин, плотно связанных внутри и слабо между собой.

![Кластеризованный граф](images/clustered_graph.svg)

**Полное количество связей в кластере** из $N$ вершин (сколько рёбер максимально возможно):

- неориентированный: $S = \dfrac{N(N-1)}{2}$ (каждая пара соединена один раз);
- ориентированный: $S = N(N-1)$ (каждая пара — в обе стороны).

**Коэффициент кластеризации** $K$ — отношение фактического числа связей к максимально возможному:

$$K = \frac{\text{фактическое число рёбер}}{S}, \qquad 0 < K \le 1.$$

Чем ближе $K$ к 1, тем плотнее сообщество (все со всеми знакомы); чем ближе к 0 — тем разреженнее.

**Взвешенный вариант.** Рёбрам можно приписать **вес** — силу связи (частота общения, число совместных действий):

![Взвешенный орграф E-A-B](images/weighted_eab.svg)

Здесь $E \to A$ с весом 3, $E \to B$ с весом 2, $B \to A$ с весом 1 — влияние E на A сильнее прямого ($3$), чем через B ($2$ и затем $1$).

## Основные параметры социального графа (2)

- **Гомофилия** — степень, в которой пользователь образует связи с **подобными** себе. Сходство может определяться по полу, возрасту, общественному положению, образовательному уровню и т. д. («подобное тянется к подобному»).
- **Множественность** (или **сила связи**) — число «множественных» связей между пользователями. Например, два человека, которые и дружат, и работают вместе, имеют множественность, равную **2**.
- **Взаимность** — степень, в которой пользователи взаимодействуют между собой и отвечают взаимностью на действия друг друга (ответил на сообщение, подписался в ответ).
- **Сетевая закрытость** (**кластеризация / сплочённость**) — степень, в которой друзья пользователя являются друзьями и **друг другу**. Коэффициент кластеризации показывает, насколько сильно вершины склонны образовывать группы (сообщества), где участники соединены между собой **гораздо плотнее**, чем с остальным графом.

## Пример расчёта сплочённости

Кластер из **6 вершин** (1–6). Часть рёбер уходит наружу, к другим сообществам (серые «хвосты») — их в расчёте сплочённости **не учитываем**, считаем только связи **внутри** группы.

![Расчёт сплочённости](images/cohesion_example.svg)

Подписи у вершин — «связи внутри | всего»: у вершины 1 это **3|4** (три внутри группы, одна наружу), у вершины 2 — **4/4** (все связи внутри), у вершины 5 — **2/2**.

**Максимум связей** для 6 вершин:

$$\text{Max} = \frac{N(N-1)}{2} = \frac{6 \cdot 5}{2} = 15.$$

**Фактических внутренних рёбер:** 8 — это $(1,2), (1,4), (1,6), (2,3), (2,5), (2,6), (3,4), (4,5)$.

**Сплочённость** (коэффициент кластеризации группы):

$$K = \frac{8}{15} \approx 0{,}53.$$

То есть реализовано чуть больше половины всех возможных связей — сообщество умеренно плотное.

## Важные типы графов

![Типы графов: полный, цикл, путь](images/graph_types.svg)

**1. Полный граф $K_n$** — граф на $n$ вершинах, у которого **все вершины попарно соединены** ребром (каждая с каждой). Число рёбер $= \dfrac{n(n-1)}{2}$.

**2. Цикл $C_n$** — граф на $n$ (различных) вершинах $v_1, \dots, v_n$, где вершина $v_i$ соединена с $v_{i-1}$ и $v_{i+1}$ (для $2 \le i \le n-1$), и, кроме того, $v_1$ соединена с $v_n$ — то есть цепочка **замыкается в кольцо**. Число рёбер $= n$.

**3. Путь $P_n$** — граф с вершинами $v_1, \dots, v_n$, где $v_i$ соединена с $v_{i-1}$ и $v_{i+1}$ (для $2 \le i \le n-1$), **без замыкания** (концы $v_1$ и $v_n$ не соединены). Число рёбер $= n-1$.

Когда речь идёт о подграфе, являющемся $P_n$, говорят о **простом пути** длины $n$: слово «простой» означает, что путь является **путём-подграфом** (вершины не повторяются). То же самое верно и для циклов.

## Маршрут в графе

- **Маршрут в графе** — чередующаяся последовательность вершин и рёбер $v_1 - u_1 - v_2 - u_2 - v_3 - \dots - v_N$, в которой любые два соседних элемента **инцидентны** (вершина и следующее ребро связаны). По рёбрам и вершинам можно проходить **повторно**.
- **Путь** в графе — последовательность вершин, в которой каждая вершина соединена со следующей ребром.
- Если $v_0 = v_N$ (начало совпадает с концом) — маршрут **замкнут**, иначе **открыт**.
- Маршрут, в котором все рёбра **попарно различны** (не повторяются), называется **цепью**. Замкнутый маршрут, являющийся цепью, называется **циклом**.

![Маршрут и граф с петлёй](images/route.svg)

**Слева** — взвешенный орграф; синим выделен маршрут $A \to C \to E \to D \to F$ (его вершины подсвечены). **Справа** — неориентированный граф с **петлёй** у вершины 1 и циклом $2\!-\!3\!-\!4\!-\!5\!-\!2$.

## Связность графа

**Граф называется связным**, если между **любыми двумя** его вершинами существует маршрут (можно добраться из любой вершины в любую).

![Связный и несвязный граф](images/connectivity.svg)

- **Слева** — связный граф: из любой вершины есть путь в любую другую.
- **Справа** — **несвязный** граф: он распадается на **компоненты связности** (здесь две — треугольник и отдельная пара). Между вершинами разных компонент маршрута нет.

**Компонента связности** — максимальный «кусок» графа, внутри которого все вершины достижимы друг из друга. Связный граф состоит из **одной** компоненты.

## Простой алгоритм определения связности графа

Разделим все вершины графа на **три разновидности** (маркеры):

- **1** — вершины, про которые **ничего не известно**;
- **2** — вершины, про которые известно, что они **достижимы** из начальной, но **ещё не обработаны**;
- **3** — вершины, которые достижимы из начальной и **уже обработаны**.

**1. Первоначальная разметка**

Пометим все вершины **маркером 1** (про них ничего не известно). Выберем любую вершину (например, первую) и пометим её **маркером 2** — она достижима сама из себя.

**2. Разметка соседних вершин**

Берём любую вершину с **маркером 2**, помечаем её **маркером 3** (обработана). Все её соседи, помеченные маркером 1, получают **маркер 2**. Повторяем, пока есть вершины с маркером 2.

**3. Завершение работы**

Когда вершин с маркером 2 не осталось, все достижимые вершины помечены маркером 3, а недостижимые — так и остались с маркером 1.

- Нужен **список вершин, не входящих** в компоненту заданной вершины → собираем все вершины с **маркером 1**.
- Нужно **проверить граф на связность** → считаем вершины с маркером 1. Если их **ноль** — граф **связный** (все достижимы из начальной).

Это, по сути, **обход графа в ширину/глубину** (BFS/DFS) от стартовой вершины.

**Пошаговая разметка на примере.** Граф из 7 вершин, стартуем с $v_0$; вершина $v_6$ ни с чем не соединена. Цвет = маркер (серый — 1, оранжевый — 2, зелёный — 3):

![Пошаговая разметка вершин](images/marking_steps.svg)

На каждом шаге берём вершину с маркером 2 → красим в 3 (обработана), а её непомеченных соседей → в 2. В итоге все достижимые из $v_0$ вершины стали зелёными (маркер 3), а $v_6$ осталась серой (маркер 1). Раз есть вершина с маркером 1 → **граф несвязный**.

Дополнительно: [algolist.ru — связность графа](http://algolist.ru/maths/graphs/linked.php)

## Дополнительная задача. Определить связность орграфа

Дан **ориентированный** взвешенный граф на 5 вершинах. Нужно определить его связность — из каких вершин достижимы какие.

![Орграф задачи](images/task_connectivity_v2.svg)

Дуги (направление важно!): $1\to2$, $1\to3$, $1\to4$, $4\to2$, $4\to3$, $5\to4$, а также **двусторонние** $1\leftrightarrow5$ и $3\leftrightarrow5$.

### Решение алгоритмом маркеров из вершины 1

В орграфе ходим **только по направлению дуг** (по исходящим рёбрам).

Исходящие рёбра: из 1 → {2, 3, 4, 5}; из 2 → {} (сток); из 3 → {5}; из 4 → {2, 3}; из 5 → {1, 3, 4}.

![Пошаговое решение из вершины 1](images/task_steps_v2.svg)

**По шагам:**

1. Старт: 1 → маркер 2.
2. Обрабатываем 1 (→3): её исходящие соседи 2, 3, 4, 5 → маркер 2.
3. Обрабатываем 2 (→3): исходящих рёбер нет.
4. Обрабатываем 3 (→3): её сосед 5 уже помечен.
5. Обрабатываем 4 (→3): её соседи 2 и 3 уже обработаны.
6. Обрабатываем 5 (→3): её соседи 1, 3, 4 уже обработаны. Вершин с маркером 2 не осталось → **стоп**.

**Итог:** все вершины **{1, 2, 3, 4, 5}** получили маркер 3.

Значит, из вершины 1 достижимы **все** остальные вершины (в том числе 5 — благодаря двусторонней дуге $1\leftrightarrow5$). 

Однако граф **не сильно связный**: например, вершина 2 — **сток** (из неё нет исходящих дуг), поэтому из 2 не добраться никуда. Сильная связность требует, чтобы из **каждой** вершины были достижимы все; здесь это не так.

## Задача поиска кратчайшего пути на графе

**Задача о кратчайшем пути** — поиск самого короткого пути (цепи) между двумя точками (вершинами) на графе, при котором **минимизируется сумма весов рёбер**, составляющих путь.

![Кратчайший путь A→F](images/route.svg)

Здесь ищется путь из $A$ в $F$. Синим выделен маршрут $A \to C \to E \to D \to F$; его «длина» — не число рёбер, а **сумма весов**:

$$2 + 3 + 4 + 11 = 20.$$

Альтернатива через $B$: $A \to B \to D \to F = 4 + 10 + 11 = 25$ — длиннее. Поэтому кратчайший путь идёт через $C, E, D$.

**Вес** ребра — это «стоимость» перехода: расстояние, время, цена, потери. Задача решается специальными алгоритмами (Дейкстры, Беллмана — Форда, A\*), а не полным перебором, потому что путей может быть очень много.

## Разновидности задачи поиска кратчайшего пути

- **В заданный пункт назначения.** Найти кратчайший путь в заданную вершину-назначение $t$ из **каждой** вершины графа (кроме $t$). Если поменять направление всех рёбер, задача сводится к задаче об **одной исходной вершине** (поиск кратчайших путей из заданной вершины во все остальные).
- **Между заданной парой вершин.** Найти кратчайший путь из вершины $u$ в вершину $v$ (одна пара «откуда → куда»).
- **Между всеми парами вершин.** Найти кратчайший путь из **каждой** вершины $u$ в **каждую** вершину $v$. Можно решать и алгоритмом для одной исходной вершины (запуская его из каждой), но обычно специальные алгоритмы (например, Флойда — Уоршелла) работают быстрее.
- **Поиск максимально длинного пути** (например, **критического пути** на сетевой диаграмме проекта). Сводится к поиску кратчайшего пути, если **инвертировать веса дуг** (взять их со знаком минус); веса при этом — длительности задач.

Дополнительно: [habr.com — задачи о путях](https://habr.com/ru/post/119158/)

## Общий обзор алгоритмов

- **Алгоритм Ли (волновой)** — находит путь между вершинами $s$ и $t$ с **минимальным числом промежуточных вершин** (рёбер). Применение: трассировка электрических соединений на кристаллах микросхем и печатных платах, поиск кратчайшего расстояния на карте в стратегических играх.
- **Алгоритм Дейкстры** — кратчайший путь от **одной** вершины до всех остальных. Работает только для графов **без рёбер отрицательного веса**.
- **Алгоритм Беллмана — Форда** — кратчайшие пути от **одной** вершины до всех остальных во взвешенном графе. Вес рёбер **может быть отрицательным** (медленнее Дейкстры).
- **Алгоритм A\*** — маршрут с наименьшей стоимостью от начальной вершины к целевой, используя поиск «по первому наилучшему совпадению» (с эвристикой-оценкой оставшегося расстояния). Быстрее Дейкстры, когда есть хорошая эвристика.
- **Алгоритм Флойда — Уоршелла** — кратчайшие пути между **всеми парами** вершин взвешенного орграфа.
- **Алгоритм Джонсона** — кратчайшие пути между **всеми парами** вершин взвешенного орграфа (эффективен на разреженных графах, допускает отрицательные веса).

Дополнительно: [habr.com — с примерами кода для каждого алгоритма](https://habr.com/ru/post/111361/)

## Алгоритм Дейкстры (Dijkstra)

Находит кратчайшие пути от одной (стартовой) вершины до всех остальных. Каждой вершине хранится **текущая лучшая оценка расстояния** от старта.

1. Присвоить **начальной** вершине вес **0**.
2. Присвоить всем остальным вершинам заведомо максимальный вес (в пределе — **бесконечность**).
3. Начиная от стартовой вершины, перебрать все **инцидентные** ей вершины и присвоить путям веса, равные весу соответствующего ребра.
4. От каждой **не стартовой** вершины последовательно строить путь к каждой из инцидентных вершин, присваивая значения так же, как на шаге 3.
5. **Суммировать** вес пути с предыдущими рёбрами (накопленное расстояние до текущей вершины + вес ребра).
6. Повторять шаги 4–5, пока не будут перебраны все рёбра графа, ведущие к искомой вершине.
7. Выбрать путь с **наименьшим суммарным весом**.

**Ключевая идея (релаксация):** если через текущую вершину до соседа получается короче, чем записано — обновляем оценку соседа. На каждом шаге «фиксируется» ближайшая ещё не обработанная вершина — её расстояние уже окончательное. Именно поэтому нужны **неотрицательные** веса: иначе позже мог бы найтись более короткий путь к уже зафиксированной вершине.

Дополнительно: [prog-cpp.ru/deikstra](https://prog-cpp.ru/deikstra/), [habr.com/ru/post/111361](https://habr.com/ru/post/111361/)

## Дейкстра по шагам (с релаксацией)

Ищем кратчайшие расстояния от вершины **A** до всех остальных. У каждой вершины храним оценку $d$ — текущее лучшее известное расстояние от старта.

![Пошаговая работа алгоритма Дейкстры](images/dijkstra_steps.svg)

**Что происходит на каждом шаге:**

1. **Инициализация:** $d(A)=0$, у всех остальных $d=\infty$.
2. **Обработали A** (берём вершину с наименьшим $d$): смотрим соседей — $d(B)=4$, $d(C)=1$. Это пока **предварительные** оценки.
3. **Обработали C** ($d=1$ — сейчас наименьшее): через C до B получается $1+2=3 < 4$ → **релаксация**: $d(B)$ улучшается $4 \to 3$. Также $d(D)=1+5=6$.
4. **Обработали B** ($d=3$): через B до D $3+1=4 < 6$ → **релаксация** $d(D)\!: 6 \to 4$. И $d(E)=3+7=10$.
5. **Обработали D** ($d=4$): через D до E $4+3=7 < 10$ → **релаксация** $d(E)\!: 10 \to 7$.
6. **Обработали E** — очередь пуста, **готово**.

**Релаксация** (красное ребро) — суть алгоритма: для ребра $u \to v$ проверяем, не короче ли путь до $v$ через $u$:

$$\text{если } d(u) + w(u,v) < d(v), \text{ то } d(v) := d(u) + w(u,v).$$

На каждом шаге из «очереди» берётся вершина с **минимальной** оценкой $d$ и **фиксируется** (зелёная) — её расстояние уже окончательное. Это работает только при **неотрицательных** весах: тогда позже нельзя найти путь короче до уже зафиксированной вершины.

**Итог:** $d(A)=0,\ d(C)=1,\ d(B)=3,\ d(D)=4,\ d(E)=7$.

## Почему можно фиксировать вершину (доказательство)

Промежуточное состояние Дейкстры на графе из 20 вершин: часть уже покрашена в зелёный.

![Доказательство Дейкстры на большом графе](images/dijkstra_proof_v2.svg)

- **Зелёная зона** — вершины с уже **окончательным** d (кратчайшее расстояние найдено).
- **Жёлтая граница** — вершины, до которых дотянулись из зелёной зоны; их d пока **предварительное**.
- **Серые** — ещё не достигнуты ($d = \infty$).

Сейчас фиксируем **v** — жёлтую вершину с **минимальным** d (здесь $d(v)=27$). Утверждение: её d уже нельзя улучшить.

**Почему.** Любой другой (ещё не учтённый) путь из $s$ в $v$ обязан выйти из зелёной зоны через какую-то **жёлтую** вершину $u$ (например, $d(u)=33$). Но:

$$\text{длина пути через } u \;=\; d(u) + (\text{неотрицательные рёбра до } v) \;\ge\; d(u) \;\ge\; d(v).$$

Первое неравенство — потому что рёбра **неотрицательные**; второе — потому что $v$ выбрана как **минимум** среди жёлтых, значит $d(u) \ge d(v)$.

Итог: обходной путь к $v$ не короче $d(v)$ → значение окончательное, вершину можно красить в зелёный. Именно здесь нужна **неотрицательность весов**: при отрицательных рёбрах «остаток пути» мог бы быть отрицательным, и оценка $\ge d(v)$ сломалась бы.

## Дополнительно. A\*

Разбор прогона на графе с вершинами 1–6 (старт — вершина **1**, $d=0$; у остальных сначала «заведомо большой» вес 10000, то есть $\infty$). Веса рёбер: $(1,2)=7$, $(1,3)=9$, $(1,6)=14$, $(2,3)=10$, $(2,4)=15$, $(3,4)=11$, $(3,6)=2$, $(4,5)=6$, $(5,6)=9$.

![Пример A* / Дейкстры на графе 1–6](images/astar_example.svg)

- **Шаги 1–2** аналогичны Дейкстре: старт $d(1)=0$, остальным $\infty$; обработали 1 → соседи $d(2)=7$, $d(3)=9$, $d(6)=14$.
- **Шаг 3.** Берём вершину с минимальным весом — это **2** ($d=7$). Смотрим её соседей: $d(3)=9$ (через 2 было бы $7+10=17$ — не улучшает), $d(4)=7+15=22$.
- **Шаг 4.** Наименьший вес теперь у **3** ($d=9$). Релаксация: $d(6)=9+2=11 < 14$ и $d(4)=9+11=20 < 22$.
- **Шаг 5.** Наименьший из оставшихся — **6** ($d=11$). Релаксация: $d(5)=11+9=20$.

**Итог** (кратчайшие расстояния от 1): $d(1)=0,\ d(2)=7,\ d(3)=9,\ d(6)=11,\ d(4)=20,\ d(5)=20$.

**Чем A\* отличается от Дейкстры.** Здесь показан «чистый» прогон Дейкстры. **A\*** добавляет к оценке $d$ ещё **эвристику** $h$ — приблизительную оценку оставшегося расстояния до цели — и на каждом шаге берёт вершину с минимальным $d + h$, а не просто $d$. Это направляет поиск в сторону цели и обычно фиксирует заметно меньше вершин. Если $h=0$, A\* превращается в Дейкстру.

## Тот же пример по шагам (от пустого графа)

Прогон Дейкстры на графе 1–6 с самого начала: сперва у всех $d=\infty$, кроме старта $d(1)=0$. Красным показаны рёбра релаксации, зелёным — зафиксированные вершины.

![Дейкстра на графе 1–6 по шагам](images/astar_steps.svg)

1. **Инициализация:** $d(1)=0$, у остальных $\infty$.
2. **Обработали 1:** соседи получают $d(2)=7$, $d(3)=9$, $d(6)=14$.
3. **Обработали 2** (минимум $d=7$): $d(4)=7+15=22$; до 3 через 2 было бы $7+10=17$ — хуже, не обновляем.
4. **Обработали 3** ($d=9$): релаксация $d(6)\!:14\to11$ и $d(4)\!:22\to20$.
5. **Обработали 6** ($d=11$): $d(5)=11+9=20$.
6. **Обработали 4** ($d=20$): до 5 через 4 было бы $26$ — без улучшений.
7. **Обработали 5** ($d=20$): очередь пуста — **готово**.

**Итог:** $d(1)=0,\ d(2)=7,\ d(3)=9,\ d(6)=11,\ d(4)=20,\ d(5)=20$.

Обрати внимание на порядок фиксации — по **возрастанию** d: $0, 7, 9, 11, 20, 20$. Каждая следующая зелёная вершина не ближе предыдущей — это и есть инвариант Дейкстры.

## Задача коммивояжёра

Есть $N$ городов, соединённых дорогами. Нужно проложить **кратчайший замкнутый маршрут**, проходящий через **каждый город ровно один раз** (и вернуться в исходный). На языке графов — найти **гамильтонов цикл минимального веса**.

Разновидности:

- **Симметричная (TSP, traveling salesman problem)** — расстояния заданы между любыми двумя городами, матрица расстояний **симметрична**: $D_{ji} = D_{ij}$ (туда и обратно одинаково).
- **Асимметричная (ATSP)** — матрица может быть **несимметричной**: $D_{ji} \ne D_{ij}$ (например, из-за односторонних дорог). В более общем случае пути между некоторыми городами могут вообще **отсутствовать**.
- **С частичным упорядочиванием (SOP, sequential ordering problem)** — требуется, чтобы город $i$ был посещён **до** города $j$ (таких условий-ограничений может быть несколько).
- **Поиск гамильтонова цикла (HCP, hamiltonian cycle problem)** — обнаружение в произвольном графе замкнутого пути, проходящего через **каждую вершину ровно один раз** (без учёта весов — просто есть ли такой цикл).

**Сложность.** В отличие от задачи о кратчайшем пути, TSP — **NP-трудная**: точного быстрого алгоритма (полиномиального) не известно, число маршрутов растёт факториально $\big((N-1)!/2\big)$. Поэтому на практике применяют приближённые и эвристические методы (ближайший сосед, 2-opt, муравьиные алгоритмы, генетические).

Отличие от эйлерова цикла: эйлеров проходит по каждому **ребру** один раз (решается легко — по чётности степеней), а гамильтонов — по каждой **вершине** один раз (вычислительно тяжело).

## Задача о ранце

**Условие.** Вор пробрался на склад, где хранятся три вещи весом **10, 20, 30 кг** и стоимостью **60, 100, 120** «нанорублей» соответственно. Вор может унести максимум **50 кг** и хочет набрать вещей на **максимальную сумму**. Отбор вещей идёт **последовательно** (каждую либо берём целиком, либо нет).

Это классическая **задача о ранце 0/1** (0/1 knapsack): для каждой вещи решение бинарное — **взять (1)** или **не взять (0)**.

**Дерево перебора решений.** Каждый уровень — решение по очередной вещи (0 — не берём, 1 — берём). Три вещи → $2^3 = 8$ листьев-вариантов. В листе — суммарная цена и вес набора:

![Дерево перебора для задачи о ранце](images/knapsack_tree.svg)

Перебор всех наборов (вес ≤ 50):

| Набор | Вес | Цена |
|---|---|---|
| — | 0 | 0 |
| {1} | 10 | 60 |
| {2} | 20 | 100 |
| {3} | 30 | 120 |
| {1,2} | 30 | 160 |
| {1,3} | 40 | 180 |
| **{2,3}** | **50** | **220** ✓ |
| {1,2,3} | 60 | перегруз |

**Оптимум — взять вещи 2 и 3:** вес ровно $20+30=50$ кг, стоимость $100+120=\mathbf{220}$ (жёлтый лист). Вариант «взять всё» (280) не проходит по весу (60 > 50).

**Сложность.** Полный перебор — $2^N$ вариантов (экспонента). На практике решают **динамическим программированием** за $O(N \cdot W)$, где $W$ — вместимость ранца. Задача о ранце — тоже NP-трудная в общем случае.

## Алгоритмы (для задачи коммивояжёра)

Три класса методов.

### 1. Точные методы

Не только находят решение, но и **доказывают**, что оно наилучшее.

- **Полный перебор** перестановок $N-1$ городов (стартовый город фиксирован) — $(N-1)!$ вариантов, где $N! = N \cdot (N-1) \cdot (N-2) \cdots 2 \cdot 1$. Растёт факториально → применим только для маленьких $N$.
- **Направленный поиск с возвратами** (backtracking) — перебор вариантов «вокруг» некоторого решения с **отсечением** путей, которые уже длиннее лучшего найденного к текущему моменту.

### 2. Эвристические методы

Обычно **существенно быстрее** точных, но **не гарантируют** оптимальности. Их результат можно взять как первое приближение.

- **Жадный алгоритм** — при выборе очередного города берёт **ближайший** ещё не посещённый (nearest neighbour).
- **Метод ветвей и границ (Литтла)** — самый эффективный из известных методов **отсечения** «неперспективных» узлов на основе анализа матрицы расстояний. Строит бинарное дерево: в каждом узле две ветви — коммивояжёр **идёт** в некоторый город или **не идёт**.
- **Метод шнурка** — геометрическая вариация жадного: города охватываются замкнутым контуром, который постепенно растягивается, стараясь пройти через все вершины с минимальным увеличением длины.
- **Скользящий перебор** — переставляет местами вершины в небольшом «окне» пути, и это окно **скользит** вдоль всего маршрута. Хороший способ **улучшить** уже найденное эвристикой решение (родственно 2-opt).

### 3. Вероятностные методы

Практически **не останавливаются** — постоянно совершают случайные изменения пути в надежде получить более короткий.

- **Метод отжига** (simulated annealing) — перестановки городов с постепенно «затухающей» интенсивностью; лучшее найденное решение всегда сохраняется.
- **Генетический алгоритм** — создаётся большая популяция маршрутов, которые «**мутируют**» и «**скрещиваются**», обмениваясь участками; выживают лучшие.

Дополнительно: [synset.com — TSP / Salesman Intro](http://synset.com/ai/ru/tsp/Salesman_Intro.html)

## Генетический алгоритм

**Эвристический** алгоритм поиска для задач оптимизации и моделирования: работает путём **случайного подбора, комбинирования и вариации** искомых параметров с использованием механизмов, аналогичных **естественному отбору** в природе.

Ключевые понятия: **хромосома** — закодированное решение (например, маршрут), **популяция** — набор решений, **пригодность (fitness)** — насколько решение хорошее (для TSP — чем короче маршрут, тем выше).

**Цикл алгоритма:**

```mermaid
flowchart TD
  A([Начало]) -->|Инициализация| B[Начальная популяция]
  B -->|Отбор| C[Родительский пул]
  C -->|Группировка в пары| D[Пары родительских хромосом]
  D -->|Скрещивание| E[Популяция потомков]
  B -->|Мутация| E
  E -->|Оценка пригодности| F[Популяция пригодных потомков]
  F -->|Замена худших хромосом| B
  F -->|Выдача решения| G([Конец])
```

**По шагам:**

1. **Инициализация** — создаём случайную начальную популяцию решений.
2. **Отбор** — выбираем лучших особей в родительский пул (чем выше пригодность, тем больше шанс).
3. **Скрещивание (кроссовер)** — пары родителей обмениваются участками, порождая потомков.
4. **Мутация** — случайные небольшие изменения в потомках (для разнообразия, чтобы не застрять в локальном оптимуме).
5. **Оценка пригодности** потомков и **замена худших** особей в популяции.
6. Цикл повторяется много поколений; когда решение перестаёт улучшаться (или исчерпан лимит) — **выдаём лучшее** найденное.

Дополнительно: [видео про генетический алгоритм](https://www.youtube.com/watch?v=SfEZSyvbj2w), [Симанков В. С. — материалы](http://masters.donntu.org/2007/kita/andrevenkova/library/3/index.htm)

## Метод ветвей и границ для задачи о ранце. Простой вариант

Тот же ранец (вместимость **50 кг**, вещи 10, 20, 30 кг). Идея метода: **ветвление** — перебираем, какую вещь добавить следующей; **граница (отсечение)** — как только суммарный вес превысит вместимость, ветку дальше не раскрываем.

![Дерево ветвей и границ для ранца](images/bnb_knapsack.svg)

Как читать дерево (в узлах — вес **добавляемой** вещи, важен **накопленный** вес по пути от корня):

- **Уровень 1** — кладём первую вещь: 10, 20 или 30 кг.
- **Уровень 2** — добавляем вторую. Накопленный вес: например, ветка 20→30 даёт $20+30=50$ кг — ровно вместимость. Такие узлы отмечены **красной чертой** (**граница**): ранец полон, дальше класть нельзя.
- **Уровень 3** — попытка добавить третью вещь везде даёт $10+20+30=60 > 50$ → **перегруз**. Эти листья **отсекаются** (✗). Узлы под «границей» (серые с крестом) тоже отсекаются сразу — они бы дали 60 кг.

**Смысл отсечения:** вместо разбора всех $3! = 6$ полных перестановок метод обрубает ветки, как только видит перегруз, и не тратит на них время. В настоящем методе Литтла отсекают ещё агрессивнее — по **оценке (границе)** наилучшего возможного результата в поддереве, а не только по грубому «перегрузу».

**Итог:** допустимые «полные» наборы (вес = 50) — это ветки $20\!\to\!30$ и $30\!\to\!20$, то есть набор **{вещь 20 кг, вещь 30 кг}** — тот же оптимум, что мы нашли перебором на слайде «Задача о ранце» (стоимость 220).

## Метод ветвей и границ для задачи о ранце. Для 2х параметров

Теперь у каждой вещи **два параметра** — вес и стоимость (обрати внимание: цены здесь **другие**, чем на прошлом примере):

- 10 кг = **100 руб.**
- 20 кг = **120 руб.**
- 30 кг = **60 руб.**
- вместимость ранца — **50 кг**, нужно унести вещей на **максимальную сумму**.

Теперь у каждого узла дерева — вес добавляемой вещи (кг) **и её стоимость** (крупное число). Отслеживаем и накопленный вес (ограничение), и накопленную стоимость (её максимизируем).

![Дерево ветвей и границ с двумя параметрами](images/bnb_knapsack2.svg)

Веса отсекаются так же, как раньше: красная черта — вес ровно 50 кг, ✗ — перегруз (60 кг). Но выбор лучшего теперь идёт **по стоимости**. Полный перебор допустимых наборов:

| Набор | Вес | Стоимость |
|---|---|---|
| {10} | 10 | 100 |
| {20} | 20 | 120 |
| {30} | 30 | 60 |
| **{10, 20}** | **30** | **220** ✓ |
| {10, 30} | 40 | 160 |
| {20, 30} | 50 | 180 |
| {10,20,30} | 60 | перегруз |

**Оптимум — взять вещи 10 кг и 20 кг:** вес всего $30$ кг, стоимость $100+120=\mathbf{220}$ руб.

**Важный вывод.** Заполнять ранец «под завязку» **не всегда выгодно**: набор {20, 30} даёт ровно 50 кг, но лишь 180 руб, а более лёгкий {10, 20} (30 кг, 20 кг ещё свободно) даёт 220 руб. С двумя параметрами нельзя жадно набивать по весу — нужно сравнивать именно **стоимость**. Красная черта (полный вес) — это ограничение, а не признак оптимума.

## Задача (методом ветвей и границ)

Дано:

| Название | Стоимость, ед. | Польза в походе |
|---|---|---|
| Книга | 600 | 50 |
| Бинокль | 5000 | 100 |
| Аптечка | 1500 | 200 |
| Ноутбук | 4000 | 10 |
| Котелок | 500 | 150 |

**Задача 1.** Отобрать вещи с **максимальной суммарной стоимостью**, не превышающей **6000 ед.** (здесь стоимость — и «вес», и то, что максимизируем; по сути задача «набрать сумму как можно ближе к 6000, не перевалив»).

**Задача 2.** Отобрать вещи с **максимальной пользой**, при суммарной стоимости не более **5000 ед.** (стоимость играет роль веса-ограничения, польза — то, что максимизируем; это классический 0/1-ранец).

### Решение задачи 1 — максимум стоимости ≤ 6000

Лучшие наборы по стоимости (не более 6000):

| Набор | Стоимость |
|---|---|
| **Аптечка + Ноутбук + Котелок** | **6000** ✓ |
| Книга + Бинокль | 5600 |
| Бинокль + Котелок | 5500 |
| Аптечка + Ноутбук | 5500 |

**Ответ:** {Аптечка, Ноутбук, Котелок} — стоимость $1500+4000+500 = \mathbf{6000}$ (ровно предел, лучше некуда).

### Решение задачи 2 — максимум пользы при стоимости ≤ 5000

Лучшие наборы по пользе (стоимость не более 5000):

| Набор | Польза | Стоимость |
|---|---|---|
| **Книга + Аптечка + Котелок** | **400** ✓ | 2600 |
| Аптечка + Котелок | 350 | 2000 |
| Книга + Аптечка | 250 | 2100 |
| Аптечка | 200 | 1500 |

**Ответ:** {Книга, Аптечка, Котелок} — польза $50+200+150 = \mathbf{400}$ при стоимости всего $2600$ (из лимита 5000).

**Наблюдения.**

- В задаче 2 дорогие, но малополезные вещи (**Бинокль** 5000/100, **Ноутбук** 4000/10) вытесняются: они «съедают» почти весь лимит стоимости, а пользы дают мало. Выигрывают дешёвые и полезные (Аптечка 1500/200, Котелок 500/150).
- Как и на прошлом слайде: гнаться за заполнением лимита «под завязку» не нужно — оптимум по задаче 2 использовал лишь 2600 из 5000.
- Метод ветвей и границ строит дерево «брать/не брать» по каждой вещи и **отсекает** ветки, где ограничение уже нарушено (стоимость превышена) или где даже лучший исход поддерева хуже уже найденного.

### Дерево решения — задача 1 (максимум стоимости ≤ 6000)

Бинарное дерево «брать / не брать» по каждой вещи (**0** — не берём, **1** — берём). В узлах — **накопленная стоимость** по пути от корня. Вещи взяты в порядке убывания стоимости (Бинокль → Ноутбук → Аптечка → Книга → Котелок), чтобы отсечения срабатывали раньше.

![Дерево решения задачи 1](images/task1_tree.svg)

- **Красный ✗** — ветка «берём», где стоимость **превысила 6000** → отсекается (например, Бинокль 5000 + Ноутбук 4000 = 9000).
- **Жёлтый лист** — оптимум: стоимость **6000**.

Путь к оптимуму: Бинокль — **нет**, Ноутбук — **да** (4000), Аптечка — **да** (5500), Книга — **нет**, Котелок — **да** (6000). То есть набор **{Ноутбук, Аптечка, Котелок} = 6000** — ровно предел, больше без перебора лимита не набрать.

### Дерево решения — задача 2 (максимум пользы при стоимости ≤ 5000)

Здесь **два числа** в узле: крупно — накопленная **польза** (её максимизируем), мелко «сХХХ» — накопленная **стоимость** (ограничение ≤ 5000). Порядок вещей — по убыванию стоимости (Бинокль → Ноутбук → Аптечка → Книга → Котелок).

![Дерево решения задачи 2](images/task2_tree.svg)

- **Красный ✗** — ветка «берём», где стоимость **превысила 5000** → отсекается (например, Бинокль 5000 + Ноутбук 4000 = 9000; Бинокль 5000 + Аптечка 1500 = 6500).
- **Жёлтый лист** — оптимум по пользе: **400** при стоимости 2600.

Путь к оптимуму: Бинокль — **нет**, Ноутбук — **нет**, Аптечка — **да** (польза 200, стоимость 1500), Книга — **да** (польза 250, стоимость 2100), Котелок — **да** (польза **400**, стоимость 2600). Набор **{Аптечка, Книга, Котелок}**.

**Сравни с задачей 1.** Дорогой Бинокль (правая ветка «берём») почти сразу упирается в отсечения — после него на всё остальное почти не остаётся лимита стоимости, а пользы он даёт мало. Поэтому максимум пользы даёт «лёгкая» тройка дешёвых полезных вещей, а вовсе не заполнение бюджета под завязку.

## Алгоритмы поиска сообществ на социальных графах

**Сообщество** — плотно связанная группа вершин (внутри связей много, наружу мало). Задача — автоматически разбить граф на такие группы.

1. **Betweenness** — алгоритм на основе коэффициента **«центральности по посредничеству»** (betweenness): для ребра считают, сколько кратчайших путей между всеми парами вершин проходит через него. Рёбра-«мосты» между сообществами имеют высокий betweenness — их последовательно удаляют, и граф распадается на сообщества (алгоритм Гирвана — Ньюмана).
2. **Fastgreedy** — метод на основе **жадной оптимизации**: сообщества постепенно объединяют так, чтобы максимально росла **модулярность** (мера качества разбиения). Быстрый.
3. **Multilevel** (Лувенский метод) — **многоуровневая** оптимизация модулярности с эвристикой: локально объединяет вершины, затем «сжимает» сообщества в супервершины и повторяет. Хорошо масштабируется на большие графы.
4. **LabelPropagation** (распространение меток) — каждой вершине присваивают метку; на каждом шаге вершина берёт **самую частую метку среди соседей**. Метки «расползаются», и одинаковые образуют сообщества. Очень быстрый, но результат может слегка меняться от запуска к запуску.
5. **Walktrap** — на основе **случайных блужданий**: идея в том, что короткие случайные блуждания **не выходят** за пределы текущего сообщества (внутри связей больше). По «застреванию» блужданий определяют границы групп.

Все эти методы решают одну задачу — **кластеризацию графа на сообщества**, но отличаются идеей (мосты / модулярность / метки / блуждания), скоростью и устойчивостью результата.

## Пример алгоритма Betweenness

Метод разработан **Ньюманом и Гирваном**.

**Центральность по посредничеству** (edge betweenness) ребра считается как количество кратчайших путей между **всеми парами вершин**, проходящих через данное ребро. Если между двумя вершинами есть $N$ кратчайших путей, вклад распределяется между ними (доля). Чем больше это значение у ребра — тем **вероятнее, что оно соединяет разные сообщества**.

![Пример betweenness: мост между сообществами](images/betweenness.svg)

Числа на рёбрах — их betweenness. **Мост** между двумя группами (красное ребро) имеет самое высокое значение — **15**.

**Почему ровно 15.** Когда две группы соединены **единственным** ребром, **все** кратчайшие пути из одного сообщества в другое обязаны пройти через него. Слева 3 вершины, справа 5 → пар «левая–правая» ровно $3 \times 5 = 15$, каждая даёт +1 мосту. Отсюда betweenness моста = **15** — заметно больше, чем у внутренних рёбер.

**Как работает алгоритм (Гирван — Ньюман):**

1. Посчитать betweenness всех рёбер.
2. **Удалить** ребро с максимальным значением (наиболее вероятный «мост»).
3. Пересчитать betweenness для оставшихся рёбер.
4. Повторять — граф постепенно распадается на сообщества.

То есть алгоритм ищет и «перерезает» связи-мосты, оставляя плотные внутренние группы.


## Нормировка данных

**Нормировка** — это корректировка значений в соответствии с некоторыми функциями преобразования, с целью сделать их более **удобными для сравнения**.

**Нормировка данных требуется**, когда несовместимость единиц измерения переменных может отразиться на результатах, и **рекомендуется**, когда итоговые отчёты можно улучшить, выразив результаты в понятных/совместимых единицах.

**Зачем это нужно.** Если признаки в разных масштабах (например, вес в граммах $0{-}5000$ и цена в рублях $0{-}10$), то «большой» по числам признак задавит «маленький» в любых расчётах, где значения складываются или сравниваются по расстоянию (кластеризация, kNN, регрессия, нейросети). Нормировка приводит их к сопоставимому диапазону.

**Типовые способы:**

- **Min-max** (в диапазон $[0, 1]$): $\;x' = \dfrac{x - x_{\min}}{x_{\max} - x_{\min}}$.
- **Стандартизация (z-оценка)** — к среднему 0 и разбросу 1: $\;x' = \dfrac{x - \bar{x}}{\sigma}$.
- **Логарифмирование** $\;x' = \ln x$ — сжимает большие значения, полезно для сильно «перекошенных» данных.

Применительно к графам это особенно важно, когда рёбрам приписаны **веса в разных единицах** (расстояние, время, стоимость) — их нужно привести к общей шкале, прежде чем сравнивать или суммировать (см. следующий слайд про нормировку весов рёбер).


## Нормировка весов рёбер графа

Когда у рёбер веса в **разных единицах** (кг, рубли, время), их нельзя напрямую складывать. Приводим всё к общей шкале — например, к диапазону **[0; 100]** (нормировочные единицы, н.е.).

Формула min-max в диапазон $[0; 100]$: $\;x' = \dfrac{x - x_{\min}}{x_{\max} - x_{\min}} \cdot 100$. При $x_{\min}=0$ это просто $x' = \dfrac{x}{x_{\max}} \cdot 100$.

- **Диапазон масс** $0\ldots50$ кг → $[0;100]$: коэффициент $\dfrac{100}{50} = 2$, то есть **1 кг = 2 н.е.**
- **Диапазон стоимости** $0\ldots220$ руб. → $[0;100]$: коэффициент $\dfrac{100}{220} \approx 0{,}45$, то есть **1 руб. ≈ 0,45 н.е.**

После нормировки веса в общих единицах можно **складывать или перемножать**. Пример:

$$20\text{ кг} + 100\text{ руб.} = 20\cdot 2 + 100\cdot 0{,}45 = 40 + 45 = \mathbf{85}\text{ н.е.}$$

Так разнородные параметры (вес и цена) сводятся к **одному числу** — общей «стоимости» ребра, по которой уже можно искать кратчайший путь, минимальное дерево и т. п. Веса подбирают коэффициентами, если один параметр важнее другого.


## Задача о четырёх красках

**Любую карту на плоскости можно раскрасить всего четырьмя красками** так, чтобы никакие две соседние области (с общей границей-линией, а не точкой) не были одного цвета.

**Связь с графами:** область → вершина, общая граница → ребро. «Раскрасить карту» = раскрасить вершины так, чтобы концы каждого ребра были разного цвета (**раскраска графа**). Минимальное число цветов — **хроматическое число**. Теорема: для любого планарного графа оно ≤ 4.

**Почему хватает четырёх (и не нужно пяти):**

![K4 рисуется, K5 — нет](images/k5_vs_k4.png)

- **4 области «все со всеми»** (граф $K_4$) на плоскости нарисовать **можно** — им и нужно 4 цвета (левая картинка).
- **5 областей «все со всеми»** (граф $K_5$) нарисовать **нельзя**: хотя бы одна связь всегда пересечёт другую (правая картинка). Значит, пятёрки взаимных соседей не бывает → пятый цвет никогда не требуется. Это факт о **непланарности $K_5$** (теорема Куратовского).

**Следствие.** У участка может быть сколько угодно соседей, но не более **3** из них могут быть **попарно соседями** и ему, и друг другу (иначе получился бы запрещённый $K_5$). Именно поэтому иногда нужно 4 краски, но пяти — никогда.

**Оговорки:** соседство считается по общей границе-линии (касание в одной точке не в счёт); каждая область — единый связный кусок. Всё это верно для плоскости/сферы; на торе (бублике) уже может понадобиться до **7** красок.

**Где применяется:** раскраска карт, распределение частот вышкам-соседям, составление расписаний, распределение регистров в компиляторах — везде, где «конфликтующие» объекты нельзя пометить одинаково.


## Планарные и непланарные графы

- **Планарный граф** — граф, который можно изобразить на плоскости **без пересечения рёбер**. Карты всегда дают планарные графы. Примеры непланарных: $K_5$ (пятёрка «все со всеми») и $K_{3,3}$ — их без пересечений нарисовать нельзя (теорема Куратовского).
- **Хроматическое число графа** — минимальное количество красок, чтобы любые две **смежные** вершины были разного цвета.

![Планарный граф, раскрашенный в 4 цвета](images/planar_coloring_v2.svg)

На рисунке — планарный граф (рёбра не пересекаются), вершины помечены числом-цветом $1{-}4$: концы каждого ребра разного цвета. Здесь хроматическое число = **4** (граф содержит $K_4$, поэтому меньше нельзя, а по теореме о четырёх красках больше и не нужно).

**Связь понятий:** для планарных графов (то есть для любых карт) хроматическое число всегда $\le 4$ — это и есть теорема о четырёх красках. Непланарным графам может потребоваться больше цветов.

## Формула Эйлера

Для **связного планарного графа** (или карты), нарисованного на плоскости без пересечений:

$$V - E + F = 2$$

- **V** — число вершин;
- **E** — число рёбер;
- **F** — число граней (областей), **включая внешнюю** бесконечную область.

### Пример — 4 области, все попарно граничат (K4)

Возьмём 4 области **A, B, C, D**, каждая граничит с каждой. В графе **вершина = область**, а **ребро = общая граница** между двумя областями. Получается граф $K_4$ (четыре вершины, все соединены). Нужно ровно 4 краски.

<table><tr><td align="center"><b>Карта</b><br><img src="images/euler_k4_map.svg" width="320"></td><td align="center"><b>Граф (вершина = область)</b><br><img src="images/euler_k4_graph.svg" width="320"></td></tr></table>

- **V = 4** — четыре области (вершины A, B, C, D);
- **E = 6** — шесть границ (каждая область с каждой: 4·3/2 = 6);
- **F = 4** — четыре грани, на которые граф делит плоскость (3 внутренних треугольника F1, F2, F3 + внешняя F4);
- $V - E + F = 4 - 6 + 4 = 2$ ✓.

**Важно:** внешнюю область всегда считаем гранью. Здесь F — это грани самого рисунка графа (куски плоскости), а не исходные области карты.

**Зачем нужна.** Из формулы Эйлера выводится $E \le 3V - 6$ для простого планарного графа. Отсюда доказывается непланарность $K_5$: у него $V = 5$, значит $3\cdot5 - 6 = 9$, а рёбер $10 > 9$ — противоречие. То есть формула Эйлера строго объясняет, почему пятёрку «все со всеми» нельзя нарисовать на плоскости.

**Связь с поверхностями:** $V - E + F = 2 - 2g$, где $g$ — число «дырок» (род). Плоскость/сфера → 2, тор → 0.
