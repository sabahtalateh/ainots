# Задача 8. Описание нечёткой переменной

**Условие.** Привести пример расчёта для G-функций «НЕ», «ОЧЕНЬ», «Более-менее», «Не очень». Построить графическое представление нечётких множеств.

**Исходные данные.** Нечёткая переменная a = «цена одной пары туфель», руб., $X = \{0;\ 1\,000\,000\}$.

Носители термов:

| Терм | Носитель, руб. |
|---|---|
| даром | 0 - 500 |
| дёшево | 1 - 4 000 |
| средне | 2 000 - 15 000 |
| дорого | 8 000 - 150 000 |
| очень дорого | 80 000 - 500 000 |


## Решение

$$\text{not\_expensive} = 1-\mu_{\text{expensive}} \qquad \text{very\_expensive} = \mu_{\text{expensive}}^2 \qquad \text{more\_or\_less\_expensive} = \sqrt{\mu_{\text{expensive}}} \qquad \text{not\_very\_expensive} = 1-\mu_{\text{expensive}}^2$$

```python
# Data taken directly from the answer table (price, mu_cheap, mu_average, mu_expensive)
rows = [
    (1000, 1.00, 0,    0   ), (2000, 0.80, 0.10, 0   ), (3000, 0.40, 0.30, 0   ),
    (4000, 0.20, 0.50, 0   ), (5000, 0,    0.70, 0.10), (6000, 0,    0.90, 0.20),
    (7000, 0,    1.00, 0.30), (8000, 0,    0.80, 0.35), (9000, 0,    0.70, 0.40),
    (10000,0,    0.60, 0.45), (11000,0,    0.50, 0.50), (12000,0,    0.30, 0.55),
    (13000,0,    0.20, 0.60), (14000,0,    0.10, 0.65), (15000,0,    0,    0.70),
    (16000,0,    0,    0.75), (17000,0,    0,    0.80), (18000,0,    0,    0.85),
    (19000,0,    0,    0.90), (20000,0,    0,    0.95), (21000,0,    0,    0.98),
    (22000,0,    0,    0.98), (23000,0,    0,    0.98), (24000,0,    0,    1.00),
    (25000,0,    0,    1.00),
]

def g_not(mu):
    return 1 - mu

def g_very(mu):
    return mu ** 2

def g_more_or_less(mu):
    return mu ** 0.5

def g_not_very(mu):
    return 1 - mu ** 2

import matplotlib.pyplot as plt

price = [r[0] for r in rows]
mu_cheap = [r[1] for r in rows]
mu_average = [r[2] for r in rows]
mu_expensive = [r[3] for r in rows]

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

ax = axes[0]
ax.plot(price, mu_cheap, "o-", label="cheap", color="#3aa03a")
ax.plot(price, mu_average, "o-", label="average", color="#e08e2b")
ax.plot(price, mu_expensive, "o-", label="expensive", color="#c0392b")
ax.set_title("Base terms")
ax.set_xlabel("price, RUB")
ax.set_ylabel("mu(x)")
ax.legend()
ax.grid(True, linestyle="--", alpha=0.4)

ax = axes[1]
ax.plot(price, mu_expensive, "o-", label="expensive (base)", color="#c0392b")
ax.plot(price, [g_not(v) for v in mu_expensive], "o-", label="not_expensive = 1-mu", color="#2f6fb0")
ax.plot(price, [g_very(v) for v in mu_expensive], "o-", label="very_expensive = mu^2", color="#7a3fb3")
ax.plot(price, [g_more_or_less(v) for v in mu_expensive], "o-", label="more_or_less_expensive = sqrt(mu)", color="#3aa03a")
ax.plot(price, [g_not_very(v) for v in mu_expensive], "o-", label="not_very_expensive = 1-mu^2", color="#e08e2b")
ax.set_title("G-functions over \"expensive\"")
ax.set_xlabel("price, RUB")
ax.set_ylabel("mu(x)")
ax.legend(fontsize=8)
ax.grid(True, linestyle="--", alpha=0.4)

fig.tight_layout()
plt.show()
```
