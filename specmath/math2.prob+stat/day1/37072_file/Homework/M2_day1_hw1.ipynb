**Задача_1.** В лотерее $1000$ билетов; из них на один билет падает выигрыш $500$ руб., на $10$ билетов — выигрыши по $100$ руб., на $50$ билетов — выигрыши по $20$ руб., на $100$ билетов — выигрыши по $5$ руб., остальные билеты невыигрышные.   Некто покупает один билет.  
Найти вероятность выиграть не менее $20$ руб.

```python
import sympy as sp

p500 = sp.Rational(1, 1000)
p100 = sp.Rational(10, 1000)
p20  = sp.Rational(50, 1000)
p5   = sp.Rational(100, 1000)

pgt20 = p20 + p100 + p500
pgt20
```
