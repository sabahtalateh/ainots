a) Найти вероятность того, что все карты – крести.  
b) Найти вероятность того, что все карты одной масти.  
б) Найти вероятность того, что был вытянут хотя бы один туз.  

```python
# a) все крести
# сначала крестей 13, всего карт 52
13/52 * 12/51 * 11/50 * 10/49
```

```python
# б) все одной масти
# вероятность вытащить 4 карты 1-й любой масти
singlesuit = 13/52 * 12/51 * 11/50 * 10/49

# мастей 4
4 * singlesuit
```

```python
# в) найдём через обратное - не вытянут ни один туз, сначала 48 не тузов
noaces = 48/52 * 47/51 * 46/50 * 45/49
1 - noaces
```
