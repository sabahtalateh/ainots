__Задача 4__

Бросается игральная кость. Пусть событие `A` - появление чётного числа, событие `B` - появление числа больше трёх. Являются ли эти события независимыми?

__Решение__:   
Очевидно, что количество четных чисел и чисел больших трех равны в половину всех чисел, отсюда $P(A) = P(B) = 1/2$.  
  
Посчитаем совместную вероятность $P(AB)$.  
Событие $AB$ соответствует выпадению одновременно четного числа и числа большего трёх (это числа $4$ и $6$, т.е. два числа из шести) $\Rightarrow P(AB) = 2/6 = 1/3$.  

$$P(A|B) = \dfrac{P(AB)}{P(B)} = \dfrac{\frac{1}{3}}{\frac{1}{2}} = \dfrac{2}{3}.$$  

$$\dfrac{2}{3} ≠ \dfrac{1}{2}$$

События не являются независимыми.

__Задача 5__

Допустим, имеется некоторая очень редкая болезнь (поражает 0.1% населения). Вы приходите к врачу, вам делают тест на эту болезнь, и тест оказывается положительным. Врач говорит вам, что этот тест верно выявляет 99% больных этой болезнью и всего лишь в 1% случаев даёт ложный положительный ответ.

_Вопрос_: какова вероятность, что вы действительно больны ей?

_Подсказка_: вновь используйте формулу Байеса с раскрытием знаменателя с помощью формулы полной вероятности.

__Решение__  

А1 - пациент болен  
P(A1) = 0.001  
A2 - пациент здоров  
P(A2) = 1 - P(A1) = 0.999  
В - тест положительный  
$P(B|A1) = 0.99$  
$P(B|A2) = 0.01$  
  
Тест правильный и вы больны:  
$P(A1|B)=P(B|A1) * P(A1) / P(B)$
  
Тест ошибочный и вы здоровы:  
$P(A2|B) = P(B|A2) * P(A2) / P(B)$  
  
Формула полной вероятности для суммы положительных тестов(правильные и ложно положительные)  
$P(B) = P(B|A1) * P(A1) + P(B|A2) * P(A2)$

```python
p_b_all = 0.99 * 0.001 + 0.01 * 0.999
p_b_all
```

```python
# тест правильный и вы больны
p_yes = (0.99 * 0.001) / p_b_all
f'{p_yes:.2%}'
```

```python
# тест ошибочный и вы здоровы
p_no = (0.01 * 0.999) / p_b_all
f'{p_no:.2%}'
```

Например есть 1000 человек. Тест ошибочный в 10 случаях и по факту болеет всего 1 человек. Ошибка у 9 из 10
